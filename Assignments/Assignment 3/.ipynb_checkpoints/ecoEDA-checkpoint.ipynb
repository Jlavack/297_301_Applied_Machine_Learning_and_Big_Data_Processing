```python
import pandas as pd

eco = pd.read_csv("04_eco_river.csv")

print("Data size:", eco.shape)
print("\nData Type Overview:")
print(eco.dtypes.value_counts())

display(eco.head())
```

```python
na_sum = eco.isna().sum().sort_values(ascending=False)
display(na_sum.to_frame("missing").head(20))

non_null = eco.count().sort_values(ascending=True)
display(non_null.to_frame("non_null").head(20))
```

```python
cat_cols = ['ecoli_risk','river','location','swqaltitude','swqlanduse',
            'climate','landcover','is_river','upstream_flag']

for c in cat_cols:
    if c in eco.columns:
        print(f"\n[{c}] Category distribution:")
        display(eco[c].value_counts(dropna=False).head(10).to_frame("count"))
```

```python
import matplotlib.pyplot as plt

for c in cat_cols:
    if c in eco.columns:
        eco[c].value_counts(dropna=False).head(10).plot(kind="bar", figsize=(6,3))
        plt.title(f"{c}")
        plt.show()
```

```python
num_cols = eco.select_dtypes(include=["int64","float64"]).columns.tolist()

eco[num_cols].describe().T.head(20)
```

As can be seen, the "Safe" level is the most common sample for the ecoli_risk variable, followed by "Danger" and "Caution," indicating that water quality at most monitoring points is safe. The river and location variables have a wide range of categories, indicating that the data covers a variety of rivers and sampling points, demonstrating good geographical representation. Variables such as swqaltitude and swqlanduse reflect topography and land use types, with lowland and rural areas predominating. Among climate types, CD is the most common, while other climate types have fewer samples. The is_river variable is mostly 1, indicating that the majority of data comes from river monitoring points. The upstream_flag variable has a low proportion of upstream samples, indicating a relatively reasonable overall distribution.

```python
import seaborn as sns

plt.figure(figsize=(10,4))
sns.histplot(eco['ecoli_measure'], bins=50, kde=True, color='#1f77b4')
plt.title("Distribution of E. coli Measure")
plt.xlabel("E. coli (cfu/100ml)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=eco['ecoli_measure'], color='#ff7f0e')
plt.title("Boxplot of E. coli Measure (detecting outliers)")
plt.show()
```

As can be seen in the first plot, E. coli concentrations for most samples are concentrated in the lower range (around 0–500 cfu/100ml), with only a few observations showing very high values. This indicates that while the water quality of most rivers is relatively safe, there are still a few areas with significant pollution. This is further confirmed by the second boxplot. The boxplot shows a large number of outliers, which are significantly higher than the normal range, indicating that some areas experienced extremely high bacterial concentrations during the monitoring period.

```python
eco['date'] = pd.to_datetime(eco['date'], errors="coerce")

eco.set_index("date")['ecoli_measure'].resample("M").mean().plot(figsize=(10,4))
plt.title("Monthly Ecoli Measure Trend")
plt.show()
```

A trend chart of average values ​​from 2001 to 2017 was plotted. As can be seen, E. coli concentrations fluctuate widely, with several significant peaks. In particular, abnormal increases occurred around 2003, 2007, and 2010, suggesting possible short-term pollution events or extreme environmental changes in these years. Overall, while values ​​fluctuate periodically, no clear long-term upward or downward trend is evident.

```python
import seaborn as sns

corr_matrix = eco[num_cols].corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix, cmap="BrBG", center=0)
plt.title("Correlation Heatmap of Numeric Features")
plt.show()
```

```python
import pandas as pd
import numpy as np

eco_num = eco.apply(lambda s: pd.to_numeric(s, errors='coerce'))

valid_cols = [c for c in eco_num.columns if eco_num[c].notna().sum() >= 2]
eco_num = eco_num[valid_cols]

if 'ecoli_measure' not in eco_num.columns or eco_num['ecoli_measure'].notna().sum() < 2:\
    raise ValueError("Column 'ecoli_measure' does not have sufficient validity values ​​to compute correlation.")
```

```python
import matplotlib.pyplot as plt

corr_ser = eco_num.corr(numeric_only=True)['ecoli_measure'].drop(labels=['ecoli_measure'], errors='ignore')
corr_ser = corr_ser.dropna()

top10_idx = corr_ser.abs().sort_values(ascending=False).head(10).index
top10 = corr_ser.loc[top10_idx].sort_values()

plt.figure(figsize=(8,6))
plt.barh(top10.index, top10.values)
plt.title("Top 10 Correlations with ecoli_measure")
plt.xlabel("Correlation Coefficient")
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("Top 10 Relevance：")
display(top10.sort_values(ascending=False).to_frame("corr"))
```

```python
pos5 = corr_ser.sort_values(ascending=False).head(5)
neg5 = corr_ser.sort_values(ascending=True).head(5)

plt.figure(figsize=(7,4))
plt.barh(neg5.index, neg5.values)
plt.title("Top 5 Negative Correlations with ecoli_measure")
plt.xlabel("Correlation Coefficient")
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

plt.figure(figsize=(7,4))
plt.barh(pos5.index, pos5.values)
plt.title("Top 5 Positive Correlations with ecoli_measure")
plt.xlabel("Correlation Coefficient")
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("positive：")
display(pos5.to_frame("corr"))
print("negative：")
display(neg5.to_frame("corr"))
```

We calculated the correlation between each column and ecoli_measure and plotted a bar chart of the top ten most correlated variables. The resulting chart shows that E. coli concentrations in water bodies increase slightly when tourist accommodation activity increases in a region, potentially reflecting the impact of human activities on water quality. Negatively correlated variables include the number of farm permits and accommodation capacity indicators in some areas, suggesting that agricultural activities or management measures in specific areas may contribute to reducing water pollution.

```python
from scipy.stats import pearsonr, spearmanr
import pandas as pd
import numpy as np

target = 'ecoli_measure'
eco_num = eco.apply(lambda s: pd.to_numeric(s, errors='coerce'))
cols = [c for c in eco_num.columns if c != target and eco_num[c].notna().sum() >= 10 and eco_num[target].notna().sum() >= 10]

rows = []
for c in cols:
    x = eco_num[c]
    y = eco_num[target]
    m = pd.concat([x, y], axis=1).dropna()
    if len(m) < 30:
        continue
    r_p, p_p = pearsonr(m[c], m[target])
    r_s, p_s = spearmanr(m[c], m[target])
    rows.append([c, r_p, p_p, r_s, p_s, len(m)])

corr_test = pd.DataFrame(rows, columns=['feature','pearson_r','pearson_p','spearman_rho','spearman_p','n']).sort_values('pearson_p')
display(corr_test.head(15))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

top_feats = corr_test.nsmallest(5, 'pearson_p')['feature'].tolist()
for f in top_feats:
    plt.figure(figsize=(5,4))
    sns.regplot(data=eco, x=f, y='ecoli_measure', lowess=True, scatter_kws={'s':10}, line_kws={'lw':2})
    plt.title(f"{f} vs ecoli_measure (LOWESS)")
    plt.tight_layout()
    plt.show()
```

The direct correlation between E. coli concentrations and tourism-related variables (such as accommodation capacity, tourist arrivals, and accommodation occupancy rates) is weak. Although the Pearson correlation coefficient for these variables is slightly positive (approximately 0.13-0.14) in statistical analysis, the trend in the graph shows that the data points are scattered and concentrated in low-value areas, indicating that pollution indicators are more significantly influenced by other factors (such as rainfall, river type, or agricultural activities).

```python
top_rivers = eco['river'].value_counts().head(10).index
g = eco[eco['river'].isin(top_rivers)].groupby('river')['ecoli_measure'].mean().sort_values(ascending=False)
g.plot(kind='bar', figsize=(8,4), title='Mean ecoli_measure by River (Top 10 rivers)')
plt.ylabel('cfu/100ml'); plt.tight_layout(); plt.show()

eco['date'] = pd.to_datetime(eco['date'], errors='coerce')
eco['month_num'] = eco['date'].dt.month
eco.groupby('month_num')['ecoli_measure'].mean().plot(kind='bar', figsize=(8,4), title='Mean ecoli_measure by Month')
plt.xlabel('Month'); plt.ylabel('cfu/100ml'); plt.tight_layout(); plt.show()
```

Average E. coli concentrations across rivers show the highest levels of pollution in the Cam and Styx Rivers, while rivers like the Waitaki River are relatively clean, indicating significant differences in pollution levels among rivers. Monthly statistics show that E. coli concentrations rise significantly in the summer (especially December) and are lowest in the winter (June–August), demonstrating a clear seasonal pattern.
