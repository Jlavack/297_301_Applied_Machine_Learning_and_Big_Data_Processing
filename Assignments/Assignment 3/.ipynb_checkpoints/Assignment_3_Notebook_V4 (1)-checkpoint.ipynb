```python
# xi correlation

def xicor2(x, y, ties="auto"):
    x = np.asarray(x).flatten()
    y = np.asarray(y).flatten()
    n = len(y)

    if len(x) != n:
        raise IndexError(f"x, y length mismatch: {len(x)}, {len(y)}")

    if ties == "auto":
        ties = len(np.unique(y)) < n
    elif not isinstance(ties, bool):
        raise ValueError(f"expected ties either \"auto\" or boolean, got {ties} ({type(ties)}) instead")
    
    y = y[np.argsort(x)]
    r = rankdata(y, method="ordinal")
    nominator = np.sum(np.abs(np.diff(r)))

    if ties:
        l = rankdata(y, method="max")
        denominator = 2 * np.sum(l * (n - l))
        nominator *= n
    else:
        denominator = np.power(n, 2) - 1
        nominator *= 3

    statistic = 1 - nominator / denominator
    p_value = norm.sf(statistic, scale=2 / 5 / np.sqrt(n))

    return statistic, p_value


def plot_and_corr(df, target, y):
    df_clean = df[[target, y]].dropna()
    result = xicor2(df_clean[target], df_clean[y])
    print(target, result)

    corr_value = result[0]  # Extract scalar correlation coefficient

    fig, ax1 = plt.subplots(figsize=(10, 5))
    ax1.plot(df_clean.index, df_clean[target], label=target, color='tab:blue')
    ax1.set_ylabel(target, color='tab:blue')
    ax1.tick_params(axis='y', labelcolor='tab:blue')

    ax2 = ax1.twinx()
    ax2.plot(df_clean.index, df_clean[y], label=y, color='tab:orange')
    ax2.set_ylabel(y, color='tab:orange')
    ax2.tick_params(axis='y', labelcolor='tab:orange')

    plt.title(f"{target} vs {y} — XiCorr: {corr_value:.3f}")
    plt.tight_layout()
    plt.show()
```

```python
def map_column_between_dataframes(source_df: pd.DataFrame, target_df: pd.DataFrame, 
                                  key_column: str, source_column: str, target_column: str = None) -> None:
    if target_column is None:
        target_column = source_column
    
    mapping_dict = dict(zip(source_df[key_column], source_df[source_column]))
    target_df[target_column] = target_df[key_column].map(mapping_dict)

def process_weather_data(df, col_name, new_col_name):
    data = df.copy()
    data.rename(columns={"Observation time UTC": "date", col_name:new_col_name}, inplace= True)
    data["date"] = pd.to_datetime(data["date"], dayfirst=False).dt.strftime('%Y-%m-%d')
    data["date"] = pd.to_datetime(data["date"], dayfirst=False)    
    return data

def get_weather_df(wind, rain, temp, start, end, station):
    weather = pd.merge(wind, temp, on="date", how="outer")
    weather = pd.merge(weather, rain, on="date", how="outer")
    weather = weather.sort_values("date").reset_index(drop=True)
    # filter to match ecoli dataframe
    weather = weather[(weather["date"] >= start) & (weather["date"] <= end)]
    # imputation using mean value
    weather = weather.fillna(weather.mean(numeric_only=True))
    weather['climate_station'] = station 
    return weather
```

```python
def crosscorr(x, y):
    correlation  = signal.correlate(x, y,'full') 
    lags = signal.correlation_lags(len(x), len(y), mode='full')

    max_corr_lag = lags[np.argmax(correlation)]
    max_corr_value = np.max(correlation)
    cor_tup = (max_corr_lag, max_corr_value)
    return cor_tup

def xicor(x, y, ties="auto"):
    x = np.asarray(x).flatten()
    y = np.asarray(y).flatten()
    n = len(y)

    if len(x) != n:
        raise IndexError(f"x, y length mismatch: {len(x)}, {len(y)}")

    if ties == "auto":
        ties = len(np.unique(y)) < n
    elif not isinstance(ties, bool):
        raise ValueError(f"expected ties either \"auto\" or boolean, got {ties} ({type(ties)}) instead")
    
    y = y[np.argsort(x)]
    r = rankdata(y, method="ordinal")
    nominator = np.sum(np.abs(np.diff(r)))

    if ties:
        l = rankdata(y, method="max")
        denominator = 2 * np.sum(l * (n - l))
        nominator *= n
    else:
        denominator = np.power(n, 2) - 1
        nominator *= 3

    statistic = 1 - nominator / denominator

    return statistic

def feature_selection(X, y, n):
    min_features_to_select = n  
    # cv = train_test_split(n_splits=5)
    estimator = RandomForestClassifier(random_state=21)

    rfecv = RFECV(estimator=estimator, min_features_to_select=min_features_to_select, cv=5, scoring = 'accuracy')
    pipeline = Pipeline([
        ('feature_selection', rfecv),
        ('regressor', estimator)
    ])
    pipeline.fit(np.array(X), np.array(y))
    rfecv.fit(np.array(X), np.array(y))
    selected_features = X.columns[rfecv.support_]

    # # Print results
    print(f"Optimal number of features: {rfecv.n_features_}")
    print(f"Selected features: {selected_features.tolist()}")

    # Optionally, plot the cross-validation scores
    plt.figure(figsize=(10, 6))
    plt.plot(range(1, len(rfecv.cv_results_['mean_test_score']) + 1), rfecv.cv_results_['mean_test_score'])
    plt.xlabel('Number of features selected')
    plt.ylabel('Cross-validation score (MSE)')
    plt.title('Recursive Feature Elimination with Cross-Validation')
    plt.show()
    return selected_features.tolist()
```

### Data Collection
The target data for this project is sourced from the Ministry for the Environment via the MfE Data Service. It includes decades of E.coli and other river quality indicator samples taken from nationwide waterways. 

The initial feature data is sourced from Earth Sciences New Zealand via the DataHub. The initial features include:
- Temperature
- Wind
- Rainfall

Each climate feature and weather station must be downloaded from ESNZ individually. There are also restrictions to license-free downloads. This has provided some limitations on the climate features. Five weather stations were selected to represent:
- Southern Inland Canterbury
- Southern Coastal Canterbury
- Mid Inland Canterbury
- Banks Peninsula
- Christchurch and North Canterbury Plains

Each climate station was assigned to its corresponding waterway. Some are more accurate than others. North Canterbury is somewhat underrepresented (but also less frequently sampled). 

The next feature data is sourced from [INSERT SOURCE NAME HERE]. This covers agricultural activity in the Canterbury region, including:
- Number of beef & dairy cattle, deer, and sheep found on Canterbury farms
- Information about year on year farming consents, including new and altered consents

### Data Wrangling
Much of the target data set included repeated site identifiers. Some have been retained, such as `river` and `location` for the purpose of combining feature data sets and feature engineering, whilst others corresponding to coordinates and such have been removed. 

The target dataset was in good condition and did not require imputation; however, some climate stations did not align with the historical target data, and thus imputation was required. In most cases, this amounted to a small number of rows (between 10-100). The exception is for wind data collected at the Lyttelton climate station, representing Banks Peninsula. There is approximately 1800 rows that had to be imputed due to missing historical data. It was deemed best to use the mean in this case, as the closest station - Christchurch Aero - was not suitable for informed imputation of the waterways on the Peninsula, due to the geographical differences of the areas (mountains versus plains); however, it is worth noting that this may bias the wind speed feature where the waterway is located on Banks Peninsula. 

Similarly, a small number of rows were not aligned during the target data set and weather data set merge, and thus imputed with the mean.

There were four years of missing live stock data, which has been imputed using the forward fill technique; however, the main issue with this data is that the measurements are provided once annually, whereas the target data is measured 1-3 per month. The data has been applied regardless, as agricultural activity is paramount when considering E.coli levels in waterways, especially in an agriculturally rich region such as Canterbury. Each year was applied to the related sample irrespective of the time of year. We acknowledge that this is a limitation in the data. 

Farm consent data came in good condition and did not require extensive cleaning. The date column had to be reworked to match the format of the original data frame for merging purposes. This was merged on both year and month, so it is fairly representative of each individual sample (though the exact same frequency was not available for consent data).

#### Feature Engineering

```python
export_slaughter = export_slaughter.sort_values('date')
agri_river = agri_river.sort_values('date')

agri_river['month_num']= pd.to_datetime(agri_river['date']).dt.month
agri_river['quarter']= pd.to_datetime(agri_river['date']).dt.quarter
agri_river['m_sin']= np.sin(2*np.pi*agri_river['month_num']/12.0)
agri_river['m_cos']= np.cos(2*np.pi*agri_river['month_num']/12.0)
```

```python
agri_river['Live_bovine_export']= pd.to_numeric(agri_river['Live_bovine_export'], errors='coerce')
agri_river['live_ovine_export']= pd.to_numeric(agri_river['live_ovine_export'],  errors='coerce')
agri_river[num_col]= pd.to_numeric(agri_river[num_col],errors='coerce')
agri_river['live_total']= agri_river['Live_bovine_export'].fillna(0) + agri_river['live_ovine_export'].fillna(0)
agri_river['slaughter_total']= agri_river[num_col]
agri_river['live_bovine_share']= agri_river['Live_bovine_export'] / agri_river['live_total'].replace(0, np.nan)
agri_river['live_ovine_share']= agri_river['live_ovine_export']  / agri_river['live_total'].replace(0, np.nan)
```

```python
agri_river[[sheep_col, dairy_col, deer_col]] = agri_river[[sheep_col, dairy_col, deer_col]].apply(pd.to_numeric, errors='coerce')
agri_river[[sheep_col, dairy_col, deer_col]] = agri_river[[sheep_col, dairy_col, deer_col]].replace(0, np.nan)
agri_river['live_per_1k_sheep']= 1000.0 * agri_river['live_total']/ agri_river[sheep_col]
agri_river['live_per_1k_dairy']= 1000.0 * agri_river['live_total']/ agri_river[dairy_col]
agri_river['live_per_1k_deer']= 1000.0 * agri_river['live_total'] / agri_river[deer_col]
agri_river['slg_per_1k_sheep']= 1000.0 * agri_river['slaughter_total']/ agri_river[sheep_col]
agri_river['slg_per_1k_dairy']= 1000.0 * agri_river['slaughter_total']/ agri_river[dairy_col]
agri_river['slg_per_1k_deer'] = 1000.0 * agri_river['slaughter_total']/ agri_river[deer_col]
```

```python
fe_cols = ['live_total','Live_bovine_export','live_ovine_export','slaughter_total','ecoli_measure']
agri_river[[c+'_lag1'for c in fe_cols]]  = agri_river[fe_cols].shift(1)
agri_river[[c+'_lag3'for c in fe_cols]]  = agri_river[fe_cols].shift(3)
agri_river[[c+'_lag12'for c in fe_cols]]  = agri_river[fe_cols].shift(12)
agri_river[[c+'_ma3'for c in fe_cols]]  = agri_river[fe_cols].rolling(3).mean()
agri_river[[c+'_ma6'for c in fe_cols]]  = agri_river[fe_cols].rolling(6).mean()
agri_river[[c+'_ma12'for c in fe_cols]]  = agri_river[fe_cols].rolling(12).mean()
```

```python
agri_river['live_total_yoy']= agri_river['live_total'].pct_change(12, fill_method=None)
agri_river['bovine_export_yoy']= agri_river['Live_bovine_export'].pct_change(12, fill_method=None)
agri_river['ovine_export_yoy']= agri_river['live_ovine_export'].pct_change(12, fill_method=None)
agri_river['slaughter_total_yoy']= agri_river['slaughter_total'].pct_change(12, fill_method=None)
agri_river['ecoli_yoy']= agri_river['ecoli_measure'].pct_change(12, fill_method=None)
```

```python
log_cols = ['live_total','Live_bovine_export','live_ovine_export','slaughter_total','ecoli_measure']
agri_river[[c+'_log1p' for c in log_cols]] = np.log1p(agri_river[log_cols])
agri_river['live_x_slg']= agri_river['live_total'] * agri_river['slaughter_total']
```

```python
new_feature_cols = [c for c in agri_river.columns if any(k in c for k in [
    'live_', 'slg_', 'slaughter_total', 'bovine_', 'ovine_',
    'month_num','quarter','m_sin','m_cos','_lag','_ma','_yoy','_log1p','live_x_slg'])] +[sheep_col, dairy_col, deer_col, 'ecoli_measure', 'date']

agri_river.loc[:, new_feature_cols].to_csv("03_agri_river_features.csv", index=False, encoding="utf-8")
```

## Weather Features
[To index](#Index)   |  [To section head](#Data-Collection-and-Wrangling)

This section of EDA has uncovered that, of the three weather features (rainfall, temperature, and wind_speed), the highest correlation with Ecoli measurement comes from a lagged rainfall feature with a shift of two and no assumption of a linear relationship.

```python
#---------------------------EXPLORATORY DATA ANALYSIS FOR WEATHER FEATURE SUBSET----------------------------------------------
weather_eda = pd.read_csv("02_weather_river.csv")
# separate features
weather_features = ["wind_speed","temperature","rainfall"]
weather_eda = weather_eda[["ecoli_measure"] + weather_features]
# plot
plt.figure(figsize=(10,5))
weather_eda[weather_features].boxplot(vert=False)
plt.title("Boxplot of Weather Features", fontsize=20)
plt.xlabel("[rainfall = mm] [temperature = degrees C] [wind_speed = km/h]")
plt.grid(False)
plt.show()
```

This plot shows that both temperature and wind speed features are approximately symmetric, excluding some outliers above the upper whisker for wind speed and some outliers above and below the upper and lower whiskers for temperature. The most symmetric  of the three features is temperature, with the fewest number of outliers and no obvious skew inside the IQR box. 

On the other hand, the rainfall feature suggests the highest amount of skew in the weather feature subset, with a significant number of outliers beyond the upper whisker. While this suggests that the feature will not be suitable to algorithms with normality assumptions, and tree based ensemble methods will be suitable to such a feature, the variability in this feature may provide greater diversity for the model to capture distinctions in the Ecoli zones that are not as visible in the more symmetric features. Correlation analysis will be able to capture the validity of this line of thought.

```python
corr_matrix = weather_eda.corr()
corr_matrix["ecoli_measure"].sort_values(ascending = False)
```

```python
# AWS heatmap code
col_names = []
col_names = weather_eda.columns
fig, ax = plt.subplots(figsize=(12,7))
ax.set_title("Correlation of Weather Features", fontsize=14)
colormap = sns.color_palette("BrBG", 10)
sns.heatmap(corr_matrix, cmap=colormap, annot=True, fmt=".2f")
_ = ax.set_yticklabels(col_names)
```

As correlation requires numeric values, the plot shows the weather features against the raw Ecoli measurement available in the dataset.

This plot shows that there is very little correlation between each of the three weather features and the measurement of Ecoli in waterways. The highest correlation with Ecoli measurement is temperature, with a very weak correlation value of 0.09. This means that there is no linear relationship with Ecoli measurement, but does not necessarily imply that it will be useless as a feature in future modelling if that modelling is nonlinear.

Xi Correlation may offer more insight as it does not assume linearity.

```python
for c in weather_eda.columns:
    if c != 'ecoli_measure':
        plot_and_corr(weather_eda, c, 'ecoli_measure')
```

The Xi Correlation plots have again indicated weak relationships between the individual weather features and the Ecoli measurements; however, these relationships do appear stronger than they were when linearity was assumed. This suggests that there may be some relationship of value between these features and the target.

Rainfall has achieved the highest Xi Correlation score of the three features at 0.15. Visually, there does appear to be much more of a relationship with rainfall than with temperature or wind which clearly follow their own unrelated pattern. What is also visible is that some Ecoli peaks occur just after some rainfall peaks (though not in all instances), so a lagged rainfall feature might outperform the current feature as is.

```python
# create lagged columns and replot
for lag in [1, 2, 3]:
    col = 'rainfall_lagged_' + str(lag) 
    weather_eda[col] = weather_eda['rainfall'].shift(lag)
    weather_eda_w_lag = weather_eda[['ecoli_measure', col]].dropna()
    plot_and_corr(weather_eda_w_lag, 'ecoli_measure', col)
```

In all instances of the lagged feature, the Xi Correlation score more than doubled, suggesting that using the lagged feature may have greater performance in future modelling. The highest score achieved through this feature engineering was 0.384 on lag = 2, edging towards moderate correlation. 

In practice, this perhaps suggests that rainfall has a slight delay in its impact on Ecoli in waterways - it may take a few days/couple of weeks for the rainfall to cause enough runoff from farms to make a noticeable difference. It could also be due to the distance between an agricultural site and the sample site.

We also see that each land type has an un-equal number of total samples. As a consequnce of the earlier plot, we know that not all catchments have an equal spread of land types, nor do they have an equal number of samples. Therefore, we want to check that this doesn't impact the way the risk category is represented in the data. i.e. we want to make sure that because alpine grass is samples most, and catchments like the Rangitira river seem to have a lot, that this isn't skewing the way ecoli risk is appearing. One may think more samples will lead to more danger results simply by merit of being sampled more. Assuming everything else is equal and a river is just as likely to have ecoli present as any other (which is not the case, but for EDA purposes we shall check). 

```python
samplesL = df_eda[['catchment_name','location']]
locations = samplesL.drop_duplicates().groupby('catchment_name').size().reset_index()
locations.columns = ['catchment', 'locations']

samplesL = samplesL.groupby('catchment_name').size().reset_index()
samplesL.columns = ['catchment', 'samples']

samplesR = df_eda[['catchment_name','river']]
rivers = samplesR.drop_duplicates().groupby('catchment_name').size().reset_index()
rivers.columns = ['catchment', 'rivers']

samplesR = samplesR.groupby('catchment_name').size().reset_index()
samplesR.columns = ['catchment', 'samples']

fig, axes = plt.subplots(1, 2, figsize=(12, 8))
sns.barplot(y = 'catchment', x = 'locations', data = locations.sort_values(by='locations', ascending=False), label = 'locations', ax=axes[0])
sns.barplot(y = 'catchment', x = 'rivers', data = rivers.sort_values(by='rivers', ascending=False), label = 'rivers', ax=axes[0])
sns.barplot(y = 'catchment', x = 'samples', data = samplesL.sort_values(by='samples', ascending=False), label = 'locations', ax=axes[1])
sns.barplot(y = 'catchment', x = 'samples', data = samplesR.sort_values(by='samples', ascending=False), label = 'rivers', ax=axes[1])
axes[1].set_ylabel("")
axes[0].set_title("Locations & rivers per catchment", size =15)
axes[1].set_title("Samples per catchment", size =15)
axes[0].legend()
axes[1].legend()
plt.tight_layout()
plt.show()
```

Ecoli risk is not uniform per catchment or land type. With Estaurine Water ways having more danger measurements and places liek Pawson's Barry's aby and Living Springs having more danger measurements. We also see that this varys a lot within each catchment and land type combination. 

```python
#first how is ecoli risk across catchments
# ecoli = df_eda[['ecoli_risk', 'catchment_name']]
ecoli = pd.crosstab(df_eda['catchment_name'], df_eda['ecoli_risk']).reset_index()
ecoli[['Caution', 'Danger', 'Safe']] = ecoli[['Caution', 'Danger', 'Safe']].div(ecoli[['Caution', 'Danger', 'Safe']].sum(axis=1), axis=0) #divide each row by row sum
ecoli = pd.melt(frame=ecoli, id_vars=["catchment_name"],)

#now ecoli risk by land type
ecoli2 = df_eda.drop(labels = ['location' ,'period'], axis=1)
ecoli2 = pd.melt(frame=ecoli2, id_vars=["catchment_name", 'ecoli_risk', 'date', 'river'],).drop_duplicates()
ecoli2 = ecoli2[(ecoli2 != 0).all(axis=1)] #otherwise each risk level occurs the same number of times, because the catchments all have a value for each land type, because 0 is included. 
ecoli2.drop(labels = ['value'], axis=1, inplace = True)
ecoli3 = pd.crosstab([ecoli2['variable']], ecoli2['ecoli_risk']).reset_index()
ecoli3[['Caution', 'Danger', 'Safe']] = ecoli3[['Caution', 'Danger', 'Safe']].div(ecoli3[['Caution', 'Danger', 'Safe']].sum(axis=1), axis=0)
ecoli3 = pd.melt(frame=ecoli3, id_vars=["variable"],)

fig, axes = plt.subplots(1, 2, figsize=(12, 8))
sns.histplot(data=ecoli, y='catchment_name', hue='ecoli_risk', weights='value', multiple='stack', stat='percent',edgecolor="white", ax=axes[0])
sns.histplot(data=ecoli3, y='variable', hue='ecoli_risk', weights='value', multiple='stack', stat='percent', edgecolor="white", ax=axes[1])
axes[0].set_ylabel("")
axes[1].set_ylabel("")
axes[0].set_title("Ecoli risk per catchment", size =15)
axes[1].set_title("Ecoli risk per land type", size =15)
# axes[0].legend()
# axes[1].legend()
plt.tight_layout()
plt.show()
```

```python
#now ecoli risk by land type
ecoli2 = df_eda.drop(labels = ['location' ,'period'], axis=1)
ecoli2 = pd.melt(frame=ecoli2, id_vars=["catchment_name", 'ecoli_risk', 'date', 'river'],).drop_duplicates()
ecoli2 = ecoli2[(ecoli2 != 0).all(axis=1)] #otherwise each risk level occurs the same number of times, because the catchments all have a value for each land type, because 0 is included. 
ecoli2.drop(labels = ['value'], axis=1, inplace = True)
ecoli2 = pd.crosstab([ecoli2['catchment_name'], ecoli2['variable']], ecoli2['ecoli_risk']).reset_index()
ecoli2[['Caution', 'Danger', 'Safe']] = ecoli2[['Caution', 'Danger', 'Safe']].div(ecoli2[['Caution', 'Danger', 'Safe']].sum(axis=1), axis=0)
ecoli2 = pd.melt(frame=ecoli2, id_vars=["catchment_name", "variable"],)
ecoli2['value']= ecoli2.groupby('catchment_name')['value'].transform(lambda x: x / x.sum())

#now ecoli risk by land type
fig, axes = plt.subplots(1, 3, figsize=(30, 16))
ax1= sns.histplot(data=ecoli2.loc[ecoli2['ecoli_risk'] == 'Safe'], y='catchment_name', hue='variable', weights='value', multiple='stack', stat='percent', edgecolor="white", ax=axes[0])
ax2= sns.histplot(data=ecoli2.loc[ecoli2['ecoli_risk'] == 'Caution'], y='catchment_name', hue='variable', weights='value', multiple='stack', stat='percent', edgecolor="white", ax=axes[1])
ax3= sns.histplot(data=ecoli2.loc[ecoli2['ecoli_risk'] == 'Danger'], y='catchment_name', hue='variable', weights='value', multiple='stack', stat='percent', edgecolor="white", ax=axes[2])

sns.move_legend(ax1, loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=4, frameon=False)
axes[1].set_ylabel("")
axes[2].set_ylabel("")
axes[1].tick_params(axis='y', which='both', length=0)
axes[2].tick_params(axis='y', which='both', length=0)

axes[0].set_title("Ecoli Safe risk per catchment land type", size =15)
axes[1].set_title("Ecoli Caution risk per catchment land type", size =15)
axes[2].set_title("Ecoli Danger risk per catchment land type", size =15)

axes[1].legend().remove()
axes[2].legend().remove()

plt.tight_layout()
plt.show()
```

## Land Cover Feature Engineering
[To index](#Index)   |  [To section head](#Exploratory-Data-Analysis-and-Feature-Engineering)  |  [To Land Cover Features](#Land-Cover-Features)

The area value has a large range across catchments. Above we used the percentage for the EDA to make it easier ot compare between categories. So for the feature engineering we will use percentages of land type within each catchment. And, because datetime can't be used in models directly, we will turn the date column into days since the beginning of measurement. 

```python
#transform each land area into a %
df = df[df['catchment_name'] != 0]
df = df.apply(lambda x: x.astype("category") if x.dtype == "O" else x) #making all non-intetgers into categories for correlatoino later

df_per = df.iloc[:,  19:].drop(labels = 'Catchment Area', axis = 1)
df_per = df_per.div(df_per.sum(axis=1), axis=0)
# df_per = pd.concat([df.iloc[:, 0:19], df_per], axis = 1)
df_per = pd.concat([df.iloc[:, 0], df_per], axis = 1)

#convert date to time for model fitting
start = df['date'][0]
df_per['days'] = pd.to_datetime(df['date'].astype(str))
df_per['days'] = (df_per['days'] - np.datetime64(start)).astype('timedelta64[s]')
df_per['days'] = (df_per['days']/86400).astype('int64') #.dtypes

#convert ecoli_risk to label
label_encoder = LabelEncoder()
df_per['ecoli_risk'] = label_encoder.fit_transform(df_per['ecoli_risk'])
```

# Feature Engineering
[To index](#Index)   |

### Feature Engineering
In this project, not only are the features engineered, but the target variable itself. Raw E.coli measures have been converted into categorical variables representing:
- Safe Levels (<= 260 E.coli per 100ml)
- Caution Levels (260 - 550 E.coli per 100ml)
- Danger Levels (> 550 E.coli per 100ml)

These levels are informed by the MfE (2003) publication recommending guidelines for E.coli levels in recreational waterways in New Zealand. 

Using domain knowledge, several other features were engineered:
- `upstream_flag`: in Canterbury, 'upstream' often means closer to agricultural zones, and further away from coastal residential zones. It may be able to detect where E.coli enters the waterway.
- `month` and `season`: the time of year represents changes in agricultural activity, weather, and environmental conditions, all of which may influence the presence of E.coli.
- `calving_flag`: dairy farms are spread throughout the landscape of Canterbury, and increased agricultural activity during calving season has the potential to increase the presence of E.coli.
- `irrigation_flag`: similar to above, the spring and summer months see an increase on the water drawn from Canterbury waterways for increased agricultural and horticultural irrigation.

```python
from scipy.stats import pearsonr, spearmanr
import pandas as pd
import numpy as np

target = 'ecoli_measure'
eco_num = eco.apply(lambda s: pd.to_numeric(s, errors='coerce'))
cols = [c for c in eco_num.columns if c != target and eco_num[c].notna().sum() >= 10 and eco_num[target].notna().sum() >= 10]

rows = []
for c in cols:
    x = eco_num[c]
    y = eco_num[target]
    m = pd.concat([x, y], axis=1).dropna()
    if len(m) < 30:
        continue
    r_p, p_p = pearsonr(m[c], m[target])
    r_s, p_s = spearmanr(m[c], m[target])
    rows.append([c, r_p, p_p, r_s, p_s, len(m)])

corr_test = pd.DataFrame(rows, columns=['feature','pearson_r','pearson_p','spearman_rho','spearman_p','n']).sort_values('pearson_p')
display(corr_test.head(15))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

top_feats = corr_test.nsmallest(5, 'pearson_p')['feature'].tolist()
for f in top_feats:
    plt.figure(figsize=(5,4))
    sns.regplot(data=eco, x=f, y='ecoli_measure', lowess=True, scatter_kws={'s':10}, line_kws={'lw':2})
    plt.title(f"{f} vs ecoli_measure (LOWESS)")
    plt.tight_layout()
    plt.show()
```

The direct correlation between E. coli concentrations and tourism-related variables (such as accommodation capacity, tourist arrivals, and accommodation occupancy rates) is weak. Although the Pearson correlation coefficient for these variables is slightly positive (approximately 0.13-0.14) in statistical analysis, the trend in the graph shows that the data points are scattered and concentrated in low-value areas, indicating that pollution indicators are more significantly influenced by other factors (such as rainfall, river type, or agricultural activities).
