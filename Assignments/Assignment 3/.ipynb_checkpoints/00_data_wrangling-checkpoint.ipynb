## Helper Functions
[To index](#Index)   |  [To section head](#Functions)

```python
def map_column_between_dataframes(source_df: pd.DataFrame, target_df: pd.DataFrame, 
                                  key_column: str, source_column: str, target_column: str = None) -> None:
    if target_column is None:
        target_column = source_column
    
    mapping_dict = dict(zip(source_df[key_column], source_df[source_column]))
    target_df[target_column] = target_df[key_column].map(mapping_dict)

def process_weather_data(df, col_name, new_col_name):
    data = df.copy()
    data.rename(columns={"Observation time UTC": "date", col_name:new_col_name}, inplace= True)
    data["date"] = pd.to_datetime(data["date"], dayfirst=False).dt.strftime('%Y-%m-%d')
    data["date"] = pd.to_datetime(data["date"], dayfirst=False)    
    return data

def get_weather_df(wind, rain, temp, start, end, station):
    weather = pd.merge(wind, temp, on="date", how="outer")
    weather = pd.merge(weather, rain, on="date", how="outer")
    weather = weather.sort_values("date").reset_index(drop=True)
    # filter to match ecoli dataframe
    weather = weather[(weather["date"] >= start) & (weather["date"] <= end)]
    # imputation using mean value
    weather = weather.fillna(weather.mean(numeric_only=True))
    weather['climate_station'] = station 
    return weather
```

```python
#-------------------------------later river data---------------------------------
df = pd.read_csv("river-water-quality-raw-observations-2013-2017.csv", low_memory=False)
df = df[df['npid'] == "ECOLI"]
df = df[df['rcid'].isin(["CCC", "ECAN"])]
df['sdate'] = pd.to_datetime(df['sdate']).dt.strftime('%Y-%m-%d')

# ECAN sampling happens in wider Canterbury whereas CCC sampling happens in Christchurch city
df["wider_canterbury"] = np.where(df["rcid"] == "ECAN","Wider Canterbury","Christchurch")

df = df[['sdate', 'wider_canterbury', 'fdval', 'nemarid', 'river', 'location', 'swqaltitude', 'swqlanduse', 'climate', 'src_of_flw', 'landcover']]

# create the target variable
conditions = [
    (df['fdval'] <= 230),
    (df['fdval'] <= 550)
]
choices = ['Safe', 'Caution']
df['ecoli_risk'] = np.select(conditions, choices, 'Danger')

# shift more important columns to the start of the df
cols_to_move = ["ecoli_risk", "sdate", "fdval"] 
df = df[cols_to_move + [col for col in df.columns if col not in cols_to_move]]

#drop null site_id, but keep the trimmings for attempted matching later
df_null = df[df['nemarid'].isnull()]
df = df.dropna(subset=['nemarid'])

#-------------------------------earlier river data---------------------------------
df2 = pd.read_csv("river-water-quality-raw-data-by-site-1975-2013.csv", low_memory=False)
# filter for ecoli samples
df2 = df2[df2['lpid'] == "ECOLI"]
# filter for canterbury (CCC is Christchurch City Council, ECAN is Environment Canterbury)
df2 = df2[df2['srcid'].isin(["CCC", "ECAN"])]
df2['sdate'] = pd.to_datetime(df2['sdate'], format='%d/%m/%Y %H:%M').dt.strftime('%Y-%m-%d')

# ECAN sampling happens in wider Canterbury whereas CCC sampling happens in Christchurch city
#alter some columns
df2["wider_canterbury"] = np.where(df2["srcid"] == "ECAN","Wider Canterbury","Christchurch")
df2['is_river'] = np.where(df2['sflag'] == 'r', 1, 0)

#trim
df2 = df2[['sdate', 'wider_canterbury', 'fdval', 'nemarid', 'river', 'location', 'is_river']]

#make target variable
conditions = [
    (df2['fdval'] <= 230),
    (df2['fdval'] <= 550)
]
choices = ['Safe', 'Caution']
df2['ecoli_risk'] = np.select(conditions, choices, 'Danger')

#re-order columns
cols_to_move = ["ecoli_risk", "sdate", "fdval"] 
df2 = df2[cols_to_move + [col for col in df2.columns if col not in cols_to_move]]

# Create a mappings between dfs... using a join creates too many duplicate rows. 
map_column_between_dataframes(df2, df, 'nemarid', 'is_river')

columns_to_map = ['swqaltitude', 'swqlanduse', 'climate', 'src_of_flw', 'landcover']
for column in columns_to_map:
    map_column_between_dataframes(df, df2, 'nemarid', column)
# #------------------------stack them into one df-------------------------
river_df = pd.concat([df, df2], ignore_index=True).sort_values(by='sdate')
river_df = river_df.drop_duplicates()
river_df.reset_index(inplace=True, drop=True)
river_df = river_df.rename(columns={"sdate": "date", "fdval":"ecoli_measure","nemarid":"site_id"})
```

```python
#consents data
consents = pd.read_csv("consents.csv")
# more readable
consents = consents.rename(columns={"Year_Month": "date_consents", "Actual_Number":"num_farm_consents","Actual_Value":"cost_farm_consents",
                                    "Actual_Floor_area":"floor_area_farm_consents", "Altered_Actual_Number":"num_farm_consents_altered",
                                    "Altered_Actual_Value":"cost_farm_consents_altered","New_Plus_Altered_Value":"total_farm_consents_cost"})
consents = consents.loc[consents['date'].between(start, end)]

#tourism data
tourism = pd.read_csv('Tourism_2 - accomodation.csv')
tourism = tourism.loc[tourism['date'].between(start, end)]

#merge
eco_features = pd.merge(consents, tourism, on='date')
eco_features['date'] = pd.to_datetime(eco_features['date'])

eco_features['year'] = eco_features['date'].dt.year.astype(int)
eco_features['month'] = eco_features['date'].dt.month_name()

#merge with river data
eco_river = pd.merge(river_df, eco_features, on = ['year', 'month'], how = 'left')
eco_river.drop('date_y', axis=1, inplace=True)
eco_river.rename(columns={'date_x': 'date'}, inplace=True)
```

## Weather Data
[To index](#Index)   |  [To section head](#Data-Collection-and-Wrangling)


```python
#Lyttelton weather
lyt_weather = get_weather_df(process_weather_data(pd.read_csv("lyt_wind.csv", usecols=["Observation time UTC", "Speed [m/s]"]), "Speed [m/s]", "wind_speed"),
                             process_weather_data(pd.read_csv("lyt_rain.csv", usecols=["Observation time UTC", "Rainfall [mm]"]), "Rainfall [mm]", "rainfall"),
                             process_weather_data(pd.read_csv("lyt_temp.csv", usecols=["Observation time UTC", "Mean Temperature [Deg C]"]), "Mean Temperature [Deg C]", "temperature"),
                             start, end, 'Lyttelton')                            
#Timaru weather
timaru_weather = get_weather_df(process_weather_data(pd.read_csv("timaru_wind.csv", usecols=["Observation time UTC", "Speed [m/s]"]), "Speed [m/s]", "wind_speed"), 
                                process_weather_data(pd.read_csv("timaru_rain.csv", usecols=["Observation time UTC", "Rainfall [mm]"]), "Rainfall [mm]", "rainfall"), 
                                process_weather_data(pd.read_csv("timaru_temp.csv", usecols=["Observation time UTC", "Mean Temperature [Deg C]"]), "Mean Temperature [Deg C]", "temperature"), 
                                start, end, 'Timaru Aero')
#Ashburton weather
ash_weather = get_weather_df(process_weather_data(pd.read_csv("ash_wind.csv", usecols=["Observation time UTC", "Speed [m/s]"]), "Speed [m/s]", "wind_speed"), 
                             process_weather_data(pd.read_csv("ash_rain.csv", usecols=["Observation time UTC", "Rainfall [mm]"]), "Rainfall [mm]", "rainfall"), 
                             process_weather_data(pd.read_csv("ash_temp.csv", usecols=["Observation time UTC", "Mean Temperature [Deg C]"]), "Mean Temperature [Deg C]", "temperature"), 
                             start, end, 'Ashburton Aero')
#Chch weather
chch_weather = get_weather_df(process_weather_data(pd.read_csv("chch_wind.csv", usecols=["Observation time UTC", "Speed [m/s]"]), "Speed [m/s]", "wind_speed"), 
                              process_weather_data(pd.read_csv("chch_rain.csv", usecols=["Observation time UTC", "Rainfall [mm]"]), "Rainfall [mm]", "rainfall"), 
                              process_weather_data(pd.read_csv("chch_temp.csv", usecols=["Observation time UTC", "Mean Temperature [Deg C]"]), "Mean Temperature [Deg C]", "temperature"), 
                              start, end, 'Christchurch Aero')
#Tekapo weather
tekapo_weather = get_weather_df(process_weather_data(pd.read_csv("tekapo_wind.csv", usecols=["Observation time UTC", "Speed [m/s]"]), "Speed [m/s]", "wind_speed"), 
                                process_weather_data(pd.read_csv("tekapo_rain_air.csv", usecols=["Observation time UTC", "Rainfall [mm]"]), "Rainfall [mm]", "rainfall"), 
                                process_weather_data(pd.read_csv("tekapo_temp_air.csv", usecols=["Observation time UTC", "Mean Temperature [Deg C]"]), "Mean Temperature [Deg C]", "temperature"), 
                                start, end, 'Lake Tekapo')
#merge with river data
weather_all = pd.concat([lyt_weather, timaru_weather, tekapo_weather, ash_weather, chch_weather], ignore_index=True)
weather_river = river_df.merge(weather_all, on=["date", "climate_station"], how="left")
weather_river["wind_speed"]  = weather_river["wind_speed"].fillna(weather_river["wind_speed"].mean())
weather_river["temperature"] = weather_river["temperature"].fillna(weather_river["temperature"].mean())
weather_river["rainfall"]    = weather_river["rainfall"].fillna(weather_river["rainfall"].mean())
```

```python
weather_river.to_csv("02_weather_river.csv", index=False, encoding="utf-8")
```
