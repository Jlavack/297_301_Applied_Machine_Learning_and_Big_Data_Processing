```python
# xi correlation
def xicor(x, y, ties="auto", p_value=False):
    x = np.asarray(x).flatten()
    y = np.asarray(y).flatten()
    n = len(y)

    if len(x) != n:
        raise IndexError(f"x, y length mismatch: {len(x)}, {len(y)}")

    if ties == "auto":
        ties = len(np.unique(y)) < n
    elif not isinstance(ties, bool):
        raise ValueError(f"expected ties either \"auto\" or boolean, got {ties} ({type(ties)}) instead")
    
    y = y[np.argsort(x)]
    r = rankdata(y, method="ordinal")
    nominator = np.sum(np.abs(np.diff(r)))

    if ties:
        l = rankdata(y, method="max")
        denominator = 2 * np.sum(l * (n - l))
        nominator *= n
    else:
        denominator = np.power(n, 2) - 1
        nominator *= 3

    statistic = 1 - nominator / denominator
    p_value = norm.sf(statistic, scale=2 / 5 / np.sqrt(n))

    if p_value is False: return statistic
    else: return statistic, p_value

def plot_and_corr(df, target, y):
    df_clean = df[[target, y]].dropna()
    result = xicor(df_clean[target], df_clean[y], p_value = True)
    # print(target, result)

    corr_value = result[0]  # Extract scalar correlation coefficient

    fig, ax1 = plt.subplots(figsize=(10, 5))
    ax1.plot(df_clean.index, df_clean[target], label=target, color='tab:blue')
    ax1.set_ylabel(target, color='tab:blue')
    ax1.tick_params(axis='y', labelcolor='tab:blue')

    ax2 = ax1.twinx()
    ax2.plot(df_clean.index, df_clean[y], label=y, color='tab:orange')
    ax2.set_ylabel(y, color='tab:orange')
    ax2.tick_params(axis='y', labelcolor='tab:orange')

    plt.title(f"{target} vs {y} — XiCorr: {corr_value:.3f}")
    plt.tight_layout()
    plt.show()
```

```python
def map_column_between_dataframes(source_df: pd.DataFrame, target_df: pd.DataFrame, 
                                  key_column: str, source_column: str, target_column: str = None) -> None:
    if target_column is None:
        target_column = source_column
    
    mapping_dict = dict(zip(source_df[key_column], source_df[source_column]))
    target_df[target_column] = target_df[key_column].map(mapping_dict)

def process_weather_data(df, col_name, new_col_name):
    data = df.copy()
    data.rename(columns={"Observation time UTC": "date", col_name:new_col_name}, inplace= True)
    data["date"] = pd.to_datetime(data["date"], dayfirst=False).dt.strftime('%Y-%m-%d')
    data["date"] = pd.to_datetime(data["date"], dayfirst=False)    
    return data

def get_weather_df(wind, rain, temp, start, end, station):
    weather = pd.merge(wind, temp, on="date", how="outer")
    weather = pd.merge(weather, rain, on="date", how="outer")
    weather = weather.sort_values("date").reset_index(drop=True)
    # filter to match ecoli dataframe
    weather = weather[(weather["date"] >= start) & (weather["date"] <= end)]
    # imputation using mean value
    weather = weather.fillna(weather.mean(numeric_only=True))
    weather['climate_station'] = station 
    return weather
```

### Data Collection
The target data for this project is sourced from the Ministry for the Environment via the MfE Data Service. It includes decades of E.coli and other river quality indicator samples taken from nationwide waterways. 

The initial feature data is sourced from Earth Sciences New Zealand via the DataHub. The initial features include:
- Temperature
- Wind
- Rainfall

Each climate feature and weather station must be downloaded from ESNZ individually. There are also restrictions to license-free downloads. This has provided some limitations on the climate features. Five weather stations were selected to represent:
- Southern Inland Canterbury
- Southern Coastal Canterbury
- Mid Inland Canterbury
- Banks Peninsula
- Christchurch and North Canterbury Plains

Each climate station was assigned to its corresponding waterway. Some are more accurate than others. North Canterbury is somewhat underrepresented (but also less frequently sampled). 

The next feature data is sourced from [INSERT SOURCE NAME HERE]. This covers agricultural activity in the Canterbury region, including:
- Number of beef & dairy cattle, deer, and sheep found on Canterbury farms
- Information about year on year farming consents, including new and altered consents

### Data Wrangling
Much of the target data set included repeated site identifiers. Some have been retained, such as `river` and `location` for the purpose of combining feature data sets and feature engineering, whilst others corresponding to coordinates and such have been removed. 

The target dataset was in good condition and did not require imputation; however, some climate stations did not align with the historical target data, and thus imputation was required. In most cases, this amounted to a small number of rows (between 10-100). The exception is for wind data collected at the Lyttelton climate station, representing Banks Peninsula. There is approximately 1800 rows that had to be imputed due to missing historical data. It was deemed best to use the mean in this case, as the closest station - Christchurch Aero - was not suitable for informed imputation of the waterways on the Peninsula, due to the geographical differences of the areas (mountains versus plains); however, it is worth noting that this may bias the wind speed feature where the waterway is located on Banks Peninsula. 

Similarly, a small number of rows were not aligned during the target data set and weather data set merge, and thus imputed with the mean.

There were four years of missing live stock data, which has been imputed using the forward fill technique; however, the main issue with this data is that the measurements are provided once annually, whereas the target data is measured 1-3 per month. The data has been applied regardless, as agricultural activity is paramount when considering E.coli levels in waterways, especially in an agriculturally rich region such as Canterbury. Each year was applied to the related sample irrespective of the time of year. We acknowledge that this is a limitation in the data. 

Farm consent data came in good condition and did not require extensive cleaning. The date column had to be reworked to match the format of the original data frame for merging purposes. This was merged on both year and month, so it is fairly representative of each individual sample (though the exact same frequency was not available for consent data).

## 2.1.1. Target Variable Feature Engineering
In this project, not only are the features engineered, but the target variable itself. Raw E.coli measures have been converted into categorical variables representing:
- Safe Levels (<= 260 E.coli per 100ml)
- Caution Levels (260 - 550 E.coli per 100ml)
- Danger Levels (> 550 E.coli per 100ml)

These levels are informed by the MfE (2003) publication recommending guidelines for E.coli levels in recreational waterways in New Zealand. 

Using domain knowledge, several other features were engineered:
- `upstream_flag`: in Canterbury, 'upstream' often means closer to agricultural zones, and further away from coastal residential zones. It may be able to detect where E.coli enters the waterway.
- `month` and `season`: the time of year represents changes in agricultural activity, weather, and environmental conditions, all of which may influence the presence of E.coli.
- `calving_flag`: dairy farms are spread throughout the landscape of Canterbury, and increased agricultural activity during calving season has the potential to increase the presence of E.coli.
- `irrigation_flag`: similar to above, the spring and summer months see an increase on the water drawn from Canterbury waterways for increased agricultural and horticultural irrigation.

## 2.3.3. Agriculture Feature Engineering
[To index](#Index)   |  [To section head](#2.-Exploratory-Data-Analysis-and-Feature-Engineering)  | [To Agriculture features](#2.3.-Agriculture-Features)

```python
#checkpoint code. uncomment and run if only doing this portion
agri_river = pd.read_csv("03_agri_river.csv").fillna(0)
agri_river['date']= pd.to_datetime(agri_river['date'])

slaughter = pd.read_csv("slaughter.csv").fillna(0)
```

```python
slaughter_cols = list(slaughter.iloc[:, 1:31].columns)

#create some species sepcific cols lists
sheep_col = agri_river[slaughter_cols].loc[:, agri_river[slaughter_cols].columns.str.contains('Sheep|Hogget|Lamb|Ram', case = False)].columns
cow_col = agri_river[slaughter_cols].loc[:, agri_river[slaughter_cols].columns.str.contains('Beef|Calves|Chopper|Bull|Steer|Veal|Heifers', case = False)].columns

agri_river['month_num']= pd.to_datetime(agri_river['date']).dt.month
agri_river['quarter']= pd.to_datetime(agri_river['date']).dt.quarter
agri_river['m_sin']= np.sin(2*np.pi*agri_river['month_num']/12.0)
agri_river['m_cos']= np.cos(2*np.pi*agri_river['month_num']/12.0)
agri_river[slaughter_cols] = agri_river[slaughter_cols].apply(pd.to_numeric, errors='coerce').fillna(0)
#create some features
agri_river['live_total']= agri_river['Live_bovine_export'].fillna(0) + agri_river['live_ovine_export'].fillna(0)
agri_river['slaughter_total']= agri_river[slaughter_cols].sum(axis=1)
agri_river['live_bovine_share']= agri_river['Live_bovine_export'] / agri_river['live_total'].replace(0, np.nan)
agri_river['live_ovine_share']= agri_river['live_ovine_export']  / agri_river['live_total'].replace(0, np.nan)

fe_cols = ['live_total','Live_bovine_export','live_ovine_export','slaughter_total','ecoli_measure']
agri_river[[c+'_lag1'for c in fe_cols]]  = agri_river[fe_cols].shift(1)
agri_river[[c+'_lag3'for c in fe_cols]]  = agri_river[fe_cols].shift(3)
agri_river[[c+'_lag12'for c in fe_cols]]  = agri_river[fe_cols].shift(12)
agri_river[[c+'_ma3'for c in fe_cols]]  = agri_river[fe_cols].rolling(3).mean()
agri_river[[c+'_ma6'for c in fe_cols]]  = agri_river[fe_cols].rolling(6).mean()
agri_river[[c+'_ma12'for c in fe_cols]]  = agri_river[fe_cols].rolling(12).mean()

agri_river['live_total_yoy']= agri_river['live_total'].pct_change(12, fill_method=None)
agri_river['bovine_export_yoy']= agri_river['Live_bovine_export'].pct_change(12, fill_method=None)
agri_river['ovine_export_yoy']= agri_river['live_ovine_export'].pct_change(12, fill_method=None)
agri_river['slaughter_total_yoy']= agri_river['slaughter_total'].pct_change(12, fill_method=None)
agri_river['ecoli_yoy']= agri_river['ecoli_measure'].pct_change(12, fill_method=None)

log_cols = ['live_total','Live_bovine_export','live_ovine_export','slaughter_total','ecoli_measure']
agri_river[[c+'_log1p' for c in log_cols]] = np.log1p(agri_river[log_cols])
agri_river['live_x_slg']= agri_river['live_total'] * agri_river['slaughter_total']

agri_river.fillna(0, inplace= True)
```

[To index](#Index)   |  [To section head](#2.-Exploratory-Data-Analysis-and-Feature-Engineering)

#### Feature Engineering
* [2.4.1 Weather Feature Engineering](#2.4.1-Weather-Feature-Engineering) 

This section of EDA has uncovered that, of the three weather features (rainfall, temperature, and wind_speed), the highest correlation with Ecoli measurement comes from a lagged rainfall feature with a shift of two and no assumption of a linear relationship.

```python
#---------------------------EXPLORATORY DATA ANALYSIS FOR WEATHER FEATURE SUBSET----------------------------------------------
weather_eda = pd.read_csv("02_weather_river.csv")
# separate features
weather_features = ["wind_speed","temperature","rainfall"]
weather_eda = weather_eda[["ecoli_measure"] + weather_features]
# plot
plt.figure(figsize=(10,5))
weather_eda[weather_features].boxplot(vert=False)
plt.title("Boxplot of Weather Features", fontsize=20)
plt.xlabel("[rainfall = mm] [temperature = degrees C] [wind_speed = km/h]")
plt.grid(False)
plt.show()
```

This plot shows that both temperature and wind speed features are approximately symmetric, excluding some outliers above the upper whisker for wind speed and some outliers above and below the upper and lower whiskers for temperature. The most symmetric  of the three features is temperature, with the fewest number of outliers and no obvious skew inside the IQR box. 

On the other hand, the rainfall feature suggests the highest amount of skew in the weather feature subset, with a significant number of outliers beyond the upper whisker. While this suggests that the feature will not be suitable to algorithms with normality assumptions, and tree based ensemble methods will be suitable to such a feature, the variability in this feature may provide greater diversity for the model to capture distinctions in the Ecoli zones that are not as visible in the more symmetric features. Correlation analysis will be able to capture the validity of this line of thought.

```python
corr_matrix = weather_eda.corr()
corr_matrix["ecoli_measure"].sort_values(ascending = False)
```

```python
# AWS heatmap code
corr_matrix = weather_eda.corr()
corr_matrix["ecoli_measure"].sort_values(ascending = False)
weather_mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

col_names = []
col_names = weather_eda.columns
fig, ax = plt.subplots(figsize=(10,5))
ax.set_title("Correlation of Weather Features", fontsize=14)
colormap = sns.color_palette("BrBG", 10)
sns.heatmap(corr_matrix, mask = weather_mask, cmap=colormap, annot=True, fmt=".2f")
_ = ax.set_yticklabels(col_names)
```

As correlation requires numeric values, the plot shows the weather features against the raw Ecoli measurement available in the dataset.

This plot shows that there is very little correlation between each of the three weather features and the measurement of Ecoli in waterways. The highest correlation with Ecoli measurement is temperature, with a very weak correlation value of 0.09. This means that there is no linear relationship with Ecoli measurement, but does not necessarily imply that it will be useless as a feature in future modelling if that modelling is nonlinear.

Xi Correlation may offer more insight as it does not assume linearity.

```python
for c in weather_eda.columns:
    if c != 'ecoli_measure':
        plot_and_corr(weather_eda, c, 'ecoli_measure')
```

The Xi Correlation plots have again indicated weak relationships between the individual weather features and the Ecoli measurements; however, these relationships do appear stronger than they were when linearity was assumed. This suggests that there may be some relationship of value between these features and the target.

Rainfall has achieved the highest Xi Correlation score of the three features at 0.15. Visually, there does appear to be much more of a relationship with rainfall than with temperature or wind which clearly follow their own unrelated pattern. What is also visible is that some Ecoli peaks occur just after some rainfall peaks (though not in all instances), so a lagged rainfall feature might outperform the current feature as is.

## 2.4.1 Weather Feature Engineering
[To index](#Index)   |  [To section head](#2.-Exploratory-Data-Analysis-and-Feature-Engineering)  |  [To Weather Features](#2.4.-Weather-Features)


In all instances of the lagged feature, the Xi Correlation score more than doubled, suggesting that using the lagged feature may have greater performance in future modelling. The highest score achieved through this feature engineering was 0.384 on lag = 2, edging towards moderate correlation. 

In practice, this perhaps suggests that rainfall has a slight delay in its impact on Ecoli in waterways - it may take a few days/couple of weeks for the rainfall to cause enough runoff from farms to make a noticeable difference. It could also be due to the distance between an agricultural site and the sample site.

```python
# create lagged columns and replot
for lag in [1, 2, 3]:
    col = 'rainfall_lagged_' + str(lag) 
    weather_eda[col] = weather_eda['rainfall'].shift(lag)
    weather_eda_w_lag = weather_eda[['ecoli_measure', col]].dropna()
    plot_and_corr(weather_eda_w_lag, 'ecoli_measure', col)
```

## 2.5.5. Land Cover Feature Engineering
[To index](#Index)   |  [To section head](#2.-Exploratory-Data-Analysis-and-Feature-Engineering)  |  [To Land Cover Features](#2.5.-Land-Cover-Features)

The area value has a large range across catchments. Above we used the percentage for the EDA to make it easier ot compare between categories. So for the feature engineering we will use percentages of land type within each catchment. And, because datetime can't be used in models directly, we will turn the date column into days since the beginning of measurement. 

```python
#transform each land area into a %
df_land = df_land[df_land['catchment_name'] != 0]
df_land = df_land.apply(lambda x: x.astype("category") if x.dtype == "O" else x) #making all non-intetgers into categories for correlatoino later

df_per = df_land.iloc[:,  19:].drop(labels = 'Catchment Area', axis = 1)
df_per = df_per.div(df_per.sum(axis=1), axis=0)
# df_per = pd.concat([df.iloc[:, 0:19], df_per], axis = 1)
df_per = pd.concat([df.iloc[:, 0], df_per], axis = 1)

#convert date to time for model fitting
start = df_land['date'][0]
df_per['days'] = pd.to_datetime(df['date'].astype(str))
df_per['days'] = (df_per['days'] - np.datetime64(start)).astype('timedelta64[s]')
df_per['days'] = (df_per['days']/86400).astype('int64') #.dtypes

#convert ecoli_risk to label
label_encoder = LabelEncoder()
df_per['ecoli_risk'] = label_encoder.fit_transform(df_per['ecoli_risk'])
```

```python
target = 'ecoli_measure'
eco_num = eco.apply(lambda s: pd.to_numeric(s, errors='coerce'))
cols = [c for c in eco_num.columns if c != target and eco_num[c].notna().sum() >= 10 and eco_num[target].notna().sum() >= 10]

rows = []
for c in cols:
    x = eco_num[c]
    y = eco_num[target]
    m = pd.concat([x, y], axis=1).dropna()
    if len(m) < 30:
        continue
    r_p, p_p = pearsonr(m[c], m[target])
    r_s, p_s = spearmanr(m[c], m[target])
    rows.append([c, r_p, p_p, r_s, p_s, len(m)])

corr_test = pd.DataFrame(rows, columns=['feature','pearson_r','pearson_p','spearman_rho','spearman_p','n']).sort_values('pearson_p')
display(corr_test.head(15))
```

#### 4.1.1. XGBoost HyperParameter Tuning
[To index](#Index)   |  [To section head](#4.-Model-Training)  |  [To XGBoost](#4.1.-XGBoost)

```python
from sklearn.model_selection import KFold, RandomizedSearchCV, cross_val_score
from sklearn.metrics import make_scorer, matthews_corrcoef
from scipy.stats import randint, uniform
from xgboost import XGBClassifier

# Define the hyperparameter search space, random or grid search or nested search
param_dist = {
    'n_estimators': randint(300, 900),
    'max_depth': randint(3, 10),
    'learning_rate': uniform(0.01, 0.5),}


# Create the model
model = XGBClassifier(
    objective='multi:softprob',
    num_class=3,
    tree_method='hist',
    random_state=42,)

# Define inner and outer cross-validation strategies
inner_cv = KFold(n_splits=5, shuffle=True, random_state=42)  # Inner loop for hyperparameter tuning
outer_cv = KFold(n_splits=3, shuffle=True, random_state=42)  # Outer loop for performance estimation

# Create a GridSearchCV object for the inner loop
random_search = RandomizedSearchCV(
    model,
    param_distributions=param_dist,
    n_iter=80,
    cv=inner_cv,
    scoring=make_scorer(matthews_corrcoef),
    random_state=21,
    error_score='raise',
    verbose=2,
    n_jobs=-1)

# Fitting the model
random_search.fit(X_train, y_train)

# Getting the best parameters
best_random_params = random_search.best_params_
print(f'Best Parameters: {best_random_params}')

# Evaluating the best model
best_estimator = random_search.best_estimator_
y_pred_test= best_estimator.predict(X_test)
random_mcc= matthews_corrcoef(y_test, y_pred_test)
print(f'MCC of Best Model: {random_mcc}')
```

```python
#---------------------CV-----------------------
# Perform nested cross-validation using cross_val_score
nested_scores = cross_val_score(
    estimator=random_search,
    X=X_train,
    y=y_train,
    cv=outer_cv,
    scoring=make_scorer(matthews_corrcoef),
    n_jobs=-1)

# Output the nested cross-validation results
print(f'Nested CV MCC: {nested_scores.mean():.3f} (+/- {nested_scores.std() * 2:.3f})')
```

```python
# ----------NOTE (DELETE THIS FROM FINAL NOTEBOOK)---------------------------------------------
# PRIOR TO SELECTING BEST MODEL, ENSURE YOU HAVE DONE A FEW ROUNDS OF INTENTIONAL CHANGES TO YOUR GRID/RANDOM SEARCH
# THIS SHOWS YOU HAVE ENGAGED IN A THOUGHTFUL AND ITERATIVE PROCESS
#--------------------BEST MODEL----------------------
# After nested CV, refit the best model on the entire training set
# The best estimator from the inner loop is automatically refit on the full training data
best_model = random_search.best_estimator_ 
best_model.fit(X_train, y_train) 
# Now use the best model to make predictions on the test set
test_predictions = best_model.predict(X_test)
test_mcc = matthews_corrcoef(y_test, test_predictions)
print(f'Test MCC: {test_mcc:.3f}')


# Fit the RandomizedSearchCV
#random_search.fit(X_train, y_train)
```

```python
# Perform 5-fold cross-validation using MCC as the scoring metric
cv_scores = cross_val_score(best_model, X_train, y_train, cv=5, scoring=make_scorer(matthews_corrcoef))

# Print the MCC scores for each fold
print("MCC scores by fold:", cv_scores)

# Print the mean MCC and standard deviation
print("Mean MCC: {:.4f} (+/- {:.4f})".format(cv_scores.mean(), cv_scores.std() * 2))
```

#### 4.3.3. CatBoost Model Calibration
[To index](#Index)   |  [To section head](#4.-Model-Training)  |  [To CatBoost](#4.3.-CatBoost)

```python
from sklearn.impute import SimpleImputer
from imblearn.over_sampling import SMOTENC, SMOTE
from sklearn.calibration import CalibratedClassifierCV, CalibrationDisplay
from sklearn.metrics import matthews_corrcoef
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

cat_cols = list(X_train.select_dtypes(include=['object', 'category']).columns)
num_cols = [c for c in X_train.columns if c not in cat_cols]

imp_num = SimpleImputer(strategy='median')
imp_cat = SimpleImputer(strategy='most_frequent')

if len(num_cols) > 0:
    X_train_num = imp_num.fit_transform(X_train[num_cols])
    X_test_num  = imp_num.transform(X_test[num_cols])
else:
    X_train_num = np.empty((len(X_train), 0))
    X_test_num  = np.empty((len(X_test), 0))

if len(cat_cols) > 0:
    X_train_cat = imp_cat.fit_transform(X_train[cat_cols])
    X_test_cat  = imp_cat.transform(X_test[cat_cols])
else:
    X_train_cat = np.empty((len(X_train), 0))
    X_test_cat  = np.empty((len(X_test), 0))

cols_order = num_cols + cat_cols
X_train_imp = pd.DataFrame(np.hstack([X_train_num, X_train_cat]), columns=cols_order, index=X_train.index)
X_test_imp  = pd.DataFrame(np.hstack([X_test_num,  X_test_cat]),  columns=cols_order, index=X_test.index)

if len(cat_cols) > 0:
    for c in cat_cols:
        X_train_imp[c] = X_train_imp[c].astype(str)
        X_test_imp[c]  = X_test_imp[c].astype(str)

if len(cat_cols) > 0:
    cat_idx = [X_train_imp.columns.get_loc(c) for c in cat_cols]
    sm = SMOTENC(categorical_features=cat_idx, random_state=42)
else:
    sm = SMOTE(random_state=42)

X_train_res, y_train_res = sm.fit_resample(X_train_imp, y_train)
print("After over-sampling:", X_train_res.shape, y_train_res.shape)

fig, ax = plt.subplots(1, 2, figsize=(10,4))
sns.countplot(x=y_train, ax=ax[0])
ax[0].set_title("Before SMOTE: Class Distribution")
sns.countplot(x=y_train_res, ax=ax[1])
ax[1].set_title("After SMOTE: Class Distribution")
plt.tight_layout()
plt.show()

cb_best = best_model.best_estimator_
cb_best.fit(
    X_train_res, y_train_res,
    cat_features=[X_train_res.columns.get_loc(c) for c in cat_cols] if len(cat_cols)>0 else None,
    verbose=False
)

y_pred = cb_best.predict(X_test_imp)
print(f"Test MCC after resampling: {matthews_corrcoef(y_test, y_pred):.3f}")


calibrated = CalibratedClassifierCV(cb_best, method='isotonic', cv=3)
calibrated.fit(X_train_res, y_train_res)
proba = calibrated.predict_proba(X_test_imp)
classes = calibrated.classes_


n_cls = len(classes)
fig, axes = plt.subplots(1, n_cls, figsize=(5*n_cls, 4), sharey=True)
if n_cls == 1:
    axes = [axes]

for i, cls in enumerate(classes):
    y_true_bin = (y_test == cls).astype(int)
    y_prob_cls = proba[:, i]
    CalibrationDisplay.from_predictions(
        y_true=y_true_bin,
        y_prob=y_prob_cls,
        n_bins=10,
        name=f"Class {cls}",
        ax=axes[i]
    )
    axes[i].set_title(f"Calibration – Class {cls}")

plt.tight_layout()
plt.show()

```

### 4.4. Random Forest
[To index](#Index)   |  [To section head](#4.-Model-Training) 

Random Forest is an ensemble of decision trees. The randomness refers to both the bootstrapping of samples, and the random selection of features at each split. 

The algorithm's bootstrapping ensures diversity of the trees, and thus the reduction of variance is innate to Random Forest. It does not have the same inbuilt protection against bias, but this can be achieved by tuning the maximum depth of the trees. There is a risk of overfitting with this method.

Whilst Random Forest does have a built in mechanism for cross validation with bootstrapping, this is not as robust as external cross validation, so this is still important in the evaluation and comparison of differently tuned models. Thus, external cross validation will be utilised throughout the tuning process. 

The scikit-learn documentation lists the following default values for Random Forest Classifier's parameters:
- n_estimators=100, 
- criterion='gini'
- max_depth=None
- min_samples_split=2
- min_samples_leaf=1
- min_weight_fraction_leaf=0.0
- max_features='sqrt'
- max_leaf_nodes=None
- min_impurity_decrease=0.0
- bootstrap=True
- oob_score=False
- n_jobs=None
- random_state=None
- verbose=0
- warm_start=False
- class_weight=None
- ccp_alpha=0.0
- max_samples=None
- monotonic_cst=None

The most impactful on reducing variance and bias are: `n_estimators`, `criterion`, `max_depth`, `min_samples_leaf`, `max_features`, and `min_samples_split` so these will be the primary focus during the searches.

For the third round, the number of iterations will be doubled to see if that can have any meaningful impact.

```python
#------------------ROUND 3-------------------------------------------------------
# Define the hyperparameter search space, random or grid search or nested search
param_dist = {
    'n_estimators': randint(200, 400),
    'max_depth': randint(30, 100),
    'max_features': ['sqrt', None, 15, 20, 25, 30, 40],
    'min_samples_split': [3]
}

# Create the model
model = RandomForestClassifier(random_state=42)


# Define inner and outer cross-validation strategies
inner_cv = KFold(n_splits=5, shuffle=True, random_state=42)  # Inner loop for hyperparameter tuning
outer_cv = KFold(n_splits=3, shuffle=True, random_state=42)  # Outer loop for performance estimation

# Create a GridSearchCV object for the inner loop
# grid_search = GridSearchCV(
   # estimator=model,
   # param_grid=param_grid,
   # cv=inner_cv,
   # scoring=make_scorer(matthews_corrcoef),
   # n_jobs=-1,
   # verbose=0
#)
# OR Perform RandomizedSearchCV with error_score='raise' 
random_search = RandomizedSearchCV(
    model,
    param_distributions=param_dist,
    n_iter=200,
    cv=inner_cv,
    scoring=make_scorer(matthews_corrcoef),
    random_state=21,
    error_score='raise',
    verbose=2,
    n_jobs =-2
)

# Fitting the model
random_search.fit(X_train, y_train)
```

```python
# Getting the best parameters
best_random_params = random_search.best_params_
print(f'Best Parameters: {best_random_params}')

# Evaluating the best model
best_random = random_search.best_estimator_
random_pred = best_random.predict(X_test)
random_mcc = matthews_corrcoef(y_test, test_predictions)
print(f'MCC of Best Model: {random_mcc}')
```

```python
experiments_df = add_experiment(experiments_df, 'RF_3rd_Round', 0.463)
```

```python
test_results = pd.DataFrame({'True': y_test, 'Prediction': test_predictions})

fig, ax = plt.subplots(figsize=(10, 5))
sns.histplot(data=test_results, multiple='dodge', binwidth=0.6, kde=True, ax=ax)
ax.set_xticks([0, 1, 2])
ax.set_title(f'Random Forest 3rd Round \n True values vs predictions \n Test MCC = {test_mcc:.3f}')
plt.show()
```

Max_depth, max_features, and n_estimators have clearly all plateaued, with no further changes to MCC from the baseline.

This means that the tuning process was not able to increase any signal beyond what the simple baseline model picked up. Any further tuning would likely hurt performance, not increase it.  

```python
#---------------------CV-----------------------
# Perform nested cross-validation using cross_val_score
best_params = random_search.best_params_
best_model = RandomForestClassifier(n_estimators=best_params['n_estimators'], max_depth= best_params['max_depth'],
                                    max_features =  best_params['max_features'],
                                    min_samples_split= best_params['min_samples_split'])
nested_scores = cross_val_score(
    estimator=best_model,
    X=X_train,
    y=y_train,
    cv=outer_cv,
    scoring=make_scorer(matthews_corrcoef),
    n_jobs=-2
)

# Output the nested cross-validation results
print(f'Nested CV MCC: {nested_scores.mean():.3f} (+/- {nested_scores.std() * 2:.3f})')
```

The nested cross validation of the final tuned model shows a slight decrease, indicating that the performance was slightly optimistic throughout the tuning process. The MCC score of 0.458 is what the model is more likely to produce in real-world deployment.

Land related features are the most relevant to the model's decisions. Whilst these change, they don't tend to change often, so values like this could be hard-coded for ease of use in deployment, and updated periodically.

Climate has proven important to the model's decision making, but the rainfall features are weak. One of these will be kept, as this is something easy to users to input.

Interestingly, tourism related data ranks fairly high on the SHAP list. The addition of tourism data was an out of the box idea, and it has seemed to work!

```python
features_to_keep = ["swqaltitude", "Landslide", "climate",  "Artificial bare surfaces", "Herbaceous Saline Vegetation",
                   "river", "RTO_Guest arrivals (number).4", "site_id", "Depleted Grassland", "rainfall"]

X_train = X_train[features_to_keep]
X_test = X_test[features_to_keep]

best_model = random_search.fit(X_train, y_train)
# Now use the best model to make predictions on the test set
test_predictions = best_model.predict(X_test)
test_mcc = matthews_corrcoef(y_test, test_predictions)
print(f'Test MCC: {test_mcc:.3f}')
```

This is very interesting - MCC has dropped from 0.470 to 0.403, but the visualisation shows the true and predicted classes to be more balanced than previous iterations of the algorithm. So, the feature reduction has resulted in a more balanced distribution of class predictions, but also a loss in accuracy. Obviously some features that were removed had some signal. 

The ~0.07 drop is not insignificant, but it's also not a horrific loss. The domain, being a public health tool, should aim to be reasonably sound in its predictions and decision making, but it is not a domain that requires absolute integrity, like predicting cancer or life expectancy. This could still be a very functional model in deployment (with a caveat to the users about the level of uncertainty). 

#### 4.4.3. Random Forest Model Calibration
[To index](#Index)   |  [To section head](#4.-Model-Training)  |  [To Random Forest](#4.4.-Random-Forest)

```python
# this only applies to 665 rows where NaN = 0 
# the model has been able to work around this but SMOTE can't
X_train = X_train.fillna(0)
X_test = X_test.fillna(0)

# Retrain best model with smoted training data
smote_model = smote_and_retrain(best_model.best_estimator_, X_train, y_train)
# Test smoted model
test_predictions = smote_model.predict(X_test)
test_mcc = matthews_corrcoef(y_test, test_predictions)
print(f'Test MCC: {test_mcc:.3f}')
```

```python
experiments_df = add_experiment(experiments_df, 'RF_After_SMOTE', 0.390)
```

It was identified during the EDA stage that there was a class imbalance. This has the ability to negatively impact the performance of the model by creating bias towards the majority class. The previous plots have shown that there is a slight overprediction for the majority class. SMOTE can artificially rebalance the dataset and in some cases improve performance.

In this case, MCC has been reduced to 0.390, the lowest score seen thus far in the random forest experimentation. This suggests that the class imbalance was not the factor limiting performance.

```python
# Calibration and threshold testing
y_proba = best_model.predict_proba(X_test)[:, 1]
calibration_plot(best_model.best_estimator_, y_test, y_proba)
```

The safe and danger class were fairly well calibrated prior to calibration, but the caution class was notably overconfident which can be seen by the orange line veering to the right of the red dotted line. For example, at MPP of 0.5, the FoP was 0.2, meaning the model predicts a 50% chance of caution, but only 20% of those predictions are correct.

Post-calibration, the caution line is much closer to the red diagonal, suggesting that calibration is significantly improved. A majority of the safe and danger classes remain well or better aligned than they were, except for where MPP is 0.8 - they lose a little bit of confidence. Overall, the calibration has certainly improved the shape of the curve across the classes. 

```python
roc_plot(best_model, X_test, y_test)
```

As a final testament to the performance of the model, the ROC plot shows that AUC is above 0.8 for two of the classes, and around 0.7 for the weaker class. This aligns with previous indicators that the model struggles most with differentiating the caution class, but is doing a reasonable job with the other two. 

Both the caution and danger classes have a similar curve - not ideal, but still creating the bend towards the top left of the graph. There is not a clear cut off corner here for either curve, but the tradeoff between TPR and FPR appears to be around 20% FPR. The curve is, of course, weaker visually, but has the same unclear cut off. It is perhaps closer to 30% or 40% for this class.
