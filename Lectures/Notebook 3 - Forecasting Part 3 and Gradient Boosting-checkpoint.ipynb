# Ensemble-based Machine Learning - Part 3 Gradient Boosting + Forecasting


### Gradient Boosting 

Gradient Boosting is a really powerful and versatile machine learning technique that is widely used for both regression and classification tasks. It builds an ensemble of weak learners just like AdaBoost, typically decision trees, in a sequential manner. Each subsequent model in the sequence is trained to correct the **errors** made by the models that came before it - but in a very different way to AdaBoost. This iterative approach allows Gradient Boosting to create a strong predictive model from a series of weak models.

#### Key Concepts of Gradient Boosting:

1. **Ensemble Learning**:
    - Ensemble learning as we have learned is a method in machine learning where multiple models are combined to improve overall performance. In Gradient Boosting, the ensemble consists of multiple weak learners (usually shallow decision trees) that work together to produce a robust predictive model.
    - The fundamental idea behind ensemble learning is that a collection of weak models can collectively make better predictions than any individual model alone, much like how a group of experts can make more accurate decisions than a single expert.

2. **Weak Learners**:
    - A weak learner is a model that performs slightly better than random guessing. In Gradient Boosting, weak learners are often decision trees.
    - While each weak learner is simplistic and may not perform well on its own, the power of Gradient Boosting lies in combining these weak learners to form a highly accurate and complex model.

3. **Sequential Learning**:
    - Unlike other ensemble methods such as Bagging (e.g., Random Forests), where models are trained independently and then combined, Gradient Boosting builds models sequentially like AdaBoost.
    - Each model in the sequence is trained to predict the residuals (errors) of the models that came before it. By focusing on correcting these errors, each new model incrementally improves the performance of the ensemble. This approach is different to AdaBoost which alters the weighting of samples.

4. **Gradient Descent**:
    - The "gradient" in Gradient Boosting refers to the gradient descent optimization technique used to minimize the loss function, which measures the difference between the model's predictions and the actual target values.
    - The gradient indicates the direction and magnitude of the steepest increase in the loss function. By moving in the opposite direction (the negative gradient), Gradient Boosting iteratively reduces the loss, thereby improving the model's predictions.

#### How Does Gradient Boosting Work?

To understand Gradient Boosting, let's break down the process into a series of steps:

1. **Initialization**:
    - The process begins by making an initial prediction for every data point. For regression tasks, this initial prediction is often the mean of the target values for regression. For classification tasks, it could be the log-odds of the class probabilities.
    - This initial prediction serves as a starting point for the model. The subsequent steps aim to refine this prediction by reducing the prediction errors.

2. **Compute Residuals**:
    - After making the initial predictions, the next step is to compute the residuals, which are the differences between the actual target values and the current predictions.
    - These residuals represent the errors made by the model and indicate how much the predictions need to be adjusted. The residuals are equivalent to the negative gradient of the loss function in the context of gradient descent.

3. **Train a Weak Learner**:
    - A weak learner, such as a decision stump in our case, is then trained to predict these residuals. The goal of the weak learner is to approximate the residuals as closely as possible. We are effectively replacing the y values now with the **residuals**!
    - By predicting the residuals, the weak learner effectively corrects the mistakes made by the previous models, thereby "boosting" the overall performance of the model. This is called boosting but it is very different to AdaBoost.

4. **Update Predictions**:
    - The model's predictions are updated by adding the predictions of the weak learner, scaled by a learning rate. The learning rate is a hyperparameter that controls the contribution of each weak learner to the final model.
    - This iterative update allows the model to gradually improve its accuracy by reducing the residuals step by step.

5. **Repeat the Process**:
    - The process of computing residuals, training a weak learner, and updating predictions is repeated for a specified number of iterations, which corresponds to the number of weak learners in the ensemble.
    - With each iteration, the model becomes more accurate as it continuously corrects its previous mistakes.

6. **Final Prediction**:
    - The final prediction is the sum of the initial prediction and all the adjustments made by the weak learners. In regression tasks, this final prediction represents the model's best estimate of the target values. In classification tasks, the final prediction can be converted into class probabilities or decisions.

## Gradient Boosting with Decision Stump and MSE/SSE: Formal Algorithmic Notation

**Given:**
- Training data: ${(x_i, y_i)}_{i=1}^{N}$, where $x_i$ is the input feature vector and $y_i$ is the target value.
- Number of iterations (weak learners): $T$
- Learning rate: $\eta$

**Initialize:**
1. Set the initial prediction for each data point:
   $
   \hat{y}_i^{(0)} = \frac{1}{N} \sum_{i=1}^{N} y_i
   $
   for $i = 1, 2, \dots, N$. Where $\hat{y}_i^{(0)}$ is essentially $\bar{y}$
   

**For** $t = 1$ to $T$:

2. Compute the negative gradients (which for MSE are equivalent to residuals):
   $
   r_i^{(t)} = y_i - \hat{y}_i^{(t-1)}
   $
   for $i = 1, 2, \dots, N$.

3. Fit a decision stump (weak learner) to the negative gradients:
   - Find the best feature $j_t$ and threshold $thresh_t$ to split the data:
     $
     (j_t, thresh_t) = \underset{j, thresh}{\text{arg min}} \ \sum_{i=1}^{N} (r_i^{(t)} - \hat{h}_t(x_i))^2
     $
   - Where $\hat{h}_t(x_i)$ is the prediction of the decision stump, defined as:
     $
     \hat{h}_t(x_i) = \begin{cases} 
     c_L^{(t)} & \text{if } x_{ij_t} \leq thresh_t \\
     c_R^{(t)} & \text{if } x_{ij_t} > thresh_t 
     \end{cases}
     $
   - Here, $c_L^{(t)}$ and $c_R^{(t)}$ are the mean negative gradients in the left and right regions, respectively, defined by the threshold $thresh_t$.

4. Update the predictions:
   $
   \hat{y}_i^{(t)} = \hat{y}_i^{(t-1)} + \eta \cdot \hat{h}_t(x_i)
   $
   for $i = 1, 2, \dots, N$.

**End For**

**Output:**
- The final prediction for each data point:
  $
  \hat{y}^{(T)}_i = \hat{y}_i^{(0)} + \eta \sum_{t=1}^{T} \hat{h}_t(x_i)
  $
  for $i = 1, 2, \dots, N$.

---

### Explanation:
- **Initialization**: The algorithm begins by predicting the mean of the target values for all data points. This serves as a baseline prediction.

- **Negative Gradient Computation**: At each iteration, negative gradients are computed. For MSE loss, these are equivalent to residuals and represent the errors that the next weak learner should correct.

- **Fitting the Weak Learner**: A decision stump (a simple decision tree with one split) is trained to minimize the Mean Squared Error (MSE) of the negative gradients. The decision stump finds the best feature and threshold to split the data and makes predictions based on the mean negative gradients in the resulting regions.

- **Prediction Update**: The predictions are updated by adding the contribution of the decision stump, scaled by the learning rate. This gradual adjustment helps the model to improve its accuracy iteratively.

- **Final Output**: After $T$ iterations, the final predictions are expressed as the sum of the initial prediction and all the adjustments made by the weak learners, scaled by the learning rate.

#### Popularity and Success of Gradient Boosting

Gradient Boosting has become one of the most popular and successful machine learning techniques, particularly in the context of structured/tabular data. It has consistently performed well in various machine learning competitions like Kaggle  and is widely used in industry applications. The popularity of Gradient Boosting can be attributed to several factors:

- **Accuracy**: Gradient Boosting models often achieve high predictive accuracy by effectively reducing bias and variance through sequential learning.
- **Flexibility**: Gradient Boosting can be applied to a wide range of tasks, including regression, classification, and ranking. It can also accommodate various loss functions, making it adaptable to different problem types.
- **Feature Importance**: Decision trees inherently provide a measure of feature importance, which can be useful for interpreting the model and understanding the underlying data.

#### Relationship to XGBoost and LightGBM

The success of Gradient Boosting has led to the development of several advanced implementations that further enhance its performance and efficiency. Two of the most popular implementations are **XGBoost** and **LightGBM**.

1. **XGBoost (Extreme Gradient Boosting)**:
    - XGBoost is an optimized and scalable implementation of Gradient Boosting that is designed to be fast and efficient, especially on large datasets. It introduces several innovations, including:
      - **Regularization**: XGBoost includes L1 and L2 regularization to prevent overfitting and improve generalization.
      - **Parallel Processing**: XGBoost leverages parallel processing to speed up the training process, making it highly efficient on large datasets.
      - **Sparsity Awareness**: XGBoost is designed to handle sparse data efficiently, which is common in many real-world datasets.

    - XGBoost has become one of the most popular tools for machine learning practitioners due to its performance, scalability, and flexibility.

2. **LightGBM (Light Gradient Boosting Machine)**:
    - LightGBM is another highly optimized implementation of Gradient Boosting, developed by Microsoft. It is designed to be faster and more memory-efficient than traditional Gradient Boosting methods. Key features of LightGBM include:
      - **Leaf-Wise Tree Growth**: Unlike other implementations that grow trees level-wise, LightGBM grows trees leaf-wise, which can lead to deeper trees and potentially better performance.
      - **Histogram-Based Splitting**: LightGBM uses a histogram-based approach for finding the best splits, which reduces computational complexity and memory usage.
      - **Support for Large Datasets**: LightGBM is capable of handling large datasets with millions of data points and features, making it suitable for big data applications.

    - LightGBM has gained popularity due to its speed, efficiency, and ability to handle large datasets with high-dimensional features.

#### Practical Considerations

While Gradient Boosting and its variants like XGBoost and LightGBM are powerful tools, they require careful tuning of hyperparameters to achieve the best performance which tends to make them difficult to sometimes tune effectively. Some key hyperparameters to consider include:

- **Number of Estimators**: The number of weak learners (trees) to be used in the model. More estimators can lead to better performance but also increase the risk of overfitting.
- **Learning Rate**: Controls how much each weak learner contributes to the final model. A smaller learning rate often requires more estimators but can lead to better generalization.
- **Max Depth**: The maximum depth of each tree. Deeper trees can capture more complex patterns but also increase the risk of overfitting.

## A Simple Worked Example of Gradient Boosting for Regression using Sum of Squared Errors (SSE)



To demonstrate how Gradient Boosting works, let's go through a simple example with a small dataset. We'll manually perform a few iterations of Gradient Boosting, showing how it learns from residuals and combines them to make predictions.

#### Problem Setup

Consider a simple regression problem with 5 data points having one feature and the target variable:

|   X   |   y   |
|-------|-------|
|   1   |   5   |
|   2   |   7   |
|   3   |   9   |
|   4   |  11   |
|   5   |  13   |

The relationship between `X` and `y` is linear: `y = 2 * X + 3`.

#### Step 1: Initial Prediction

The first step in Gradient Boosting is to make an initial prediction. In regression, we typically start with the mean of the target values as the initial prediction.

$
\text{Initial prediction} = \hat{y}_0 = \frac{1}{N} \sum_{i=1}^{N} y_i = \frac{5 + 7 + 9 + 11 + 13}{5} = 9
$

So, the initial predictions for all data points are:

$
\hat{y}_0 = [9, 9, 9, 9, 9]
$

#### Step 2: Calculate Residuals

Next, we calculate the residuals, which are the differences between the actual values $y$ and the initial predictions $\hat{y}_0$:

$
\text{Residuals} (\bar r_0) = y - \hat{y}_0 = [5-9, 7-9, 9-9, 11-9, 13-9] = [-4, -2, 0, 2, 4]
$

|   X   |   $r_0$   |
|-------|-------|
|   1   |   -4   |
|   2   |   -2   |
|   3   |   0   |
|   4   |  2   |
|   5   |  4   |

#### Step 3: Train the First Weak Learner

A weak learner (such as a decision stump) is trained to predict the residuals and below are two example calulations with the 3.5 being the optimal:

#### Threshold \(1.5\)

**Left region**

- Uses $(X \le 1.5)$  
- Residuals $(\{-4\})  $
- Mean $(\bar r_L = -4) $ 
- SSE $= (-4 - \bar r_L)^2 = (-4 + 4)^2 = 0$

**Right region**

- Uses $(X > 1.5)$  
- Residuals $({-2, 0, 2, 4})$  
- Mean $(\bar r_R = \frac{-2 + 0 + 2 + 4}{4} = 1)$  
- SSE $= (-2 - 1)^2 + (0 - 1)^2 + (2 - 1)^2 + (4 - 1)^2 = 9 + 1 + 1 + 9 = 20)$

**Total**

- Leaf predictions $(-4, 1)$  
- Total SSE $= 0 + 20 = 20$

#### Threshold \(3.5\)

**Left region**

- Uses $(X \le 3.5)$  
- Residuals $({-4, -2, 0})$  
- Mean $\bar r_L = \frac{-4 - 2 + 0}{3} = -2 $ 
- SSE $= (-4 + 2)^2 + (-2 + 2)^2 + (0 + 2)^2 = 4 + 0 + 4 = 8$

**Right region**

- Uses $(X > 3.5)$  
- Residuals $({2, 4})$  
- Mean $(\bar r_R = \frac{2 + 4}{2} = 3$  
- SSE $= (2 - 3)^2 + (4 - 3)^2 = 1 + 1 = 2$

**Total**

- Leaf predictions (-2, 3)$
- Total SSE $= 8 + 2 = 10)$





For simplicity, let's **assume** our first weak learner predicts the following values:


$
\hat{h}_1(X) = \begin{cases} 
-2 & \text{if } X \leq 3.5 \\
3 & \text{if } X > 3.5 
\end{cases}
$

So, the predictions from the first weak learner are:

$
\hat{h}_1 = [-2, -2, -2, 3, 3]
$

#### Step 4: Update Predictions

The model's predictions are updated by adding the weak learner's predictions to the current predictions. Assuming a learning rate of `0.5`, the updated predictions are:

$
\hat{y}_1 = \hat{y}_0 + 0.5 \times \hat{h}_1 = [9, 9, 9, 9, 9] + 0.5 \times [-2, -2, -2, 3, 3] = [8, 8, 8, 10.5, 10.5]
$



#### Step 5: Calculate New Residuals

The residuals after the first iteration are recalculated:

$
\text{Residuals}  (\bar r_1) = y - \hat{y}_1 = [5-8, 7-8, 9-8, 11-10.5, 13-10.5] = [-3, -1, 1, 0.5, 2.5]
$

|   X   |   $r_1$   |
|-------|-------|
|   1   |   -3   |
|   2   |   -1   |
|   3   |   1   |
|   4   |  0.5   |
|   5   |  2.5   |

#### Step 6: Train the Second Weak Learner

The second weak learner is now trained on these new residuals. Again, for simplicity, let's **assume** it produces the following predictions (for illustration purposes):

$
\hat{h}_2(X) = \begin{cases} 
-2 & \text{if } X \leq 2.5 \\
1.3 & \text{if } X > 2.5 
\end{cases}
$

So, the predictions from the second weak learner are:

$
\hat{h}_2 = [-2, -2, 1.3, 1.3, 1.3]
$

**Check of SSE with rounded values:**

- **Left region** ($X \le 2.5$, residuals $-3, -1$):  
  Mean = $-2$  
  SSE = $(-3 - (-2))^2 + (-1 - (-2))^2 = 1 + 1 = 2$

- **Right region** ($X > 2.5$, residuals $1, 0.5, 2.5$):  
  Mean = $1.3$  
  SSE = $(1 - 1.3)^2 + (0.5 - 1.3)^2 + (2.5 - 1.3)^2 = 0.09 + 0.64 + 1.44 = 2.17$

**Total SSE** = $2 + 2.17 = 4.17$ 


#### Step 7: Update Predictions Again

We update the predictions using the second weak learner:

$
\hat{y}_2 = \hat{y}_1 + 0.5 \times \hat{h}_2
= [8, 8, 8, 10.5, 10.5] + 0.5 \times [-2, -2, 1.3, 1.3, 1.3]
= [7,\ 7,\ 8.65,\ 11.15,\ 11.15].
$

Rounded to one decimal place: $[7.0,\ 7.0,\ 8.7,\ 11.2,\ 11.2]$.

#### Step 8: Final Prediction

After these two learners, the final predictions are:

$
\hat{y} = [7,\ 7,\ 8.65,\ 11.15,\ 11.15]
$

Rounded to one decimal place: $[7.0,\ 7.0,\ 8.7,\ 11.2,\ 11.2]$.

### Summary

- **Initial Prediction**: We started with a simple guess (the mean of the target values).
- **Residuals**: At each step, we calculated how far off our predictions were from the actual values.
- **Weak Learners**: Each weak learner was trained to predict the residuals (the errors).
- **Update Predictions**: We combined the predictions from each weak learner to update our overall predictions.

### Key Takeaways

1. **Iterative Process**: Gradient Boosting builds the model iteratively, improving the predictions with each step by focusing on the residuals (errors) from the previous steps.
2. **Learning Rate**: The learning rate controls how much each weak learner's prediction influences the overall model. A smaller learning rate requires more iterations but can lead to better generalization.
3. **Weak Learners**: The weak learners are simple models that focus on capturing the residuals. By combining many of these simple models, Gradient Boosting constructs a strong predictive model.

### Implementing a Simplified Version of Gradient Boosting

In this notebook and the following exercises, we will implement a simplified version of Gradient Boosting to give us a good grasp of the algorithm. Our implementation will use a decision stump (a decision tree with only one split) as the weak learner, and we will focus on minimizing the Mean Squared Error (MSE) as the loss function.

#### Simplified Approach: Decision Stump and MSE/SSE

Recall that a decision stump is one of the simplest forms of a decision tree making it an ideal candidate for a weak learner in this context. By using a decision stump we reduce the complexity of the model allowing us to clearly observe how each step in the Gradient Boosting process contributes to the overall model and what the model actually looks like. The decision stump will split the data based on a single feature and threshold, resulting in two regions, each with its own predicted value which will be the mean of all its associated *y* values. In the context of a regression problem, the goal of the decision stump is to partition the data into two distinct regions by selecting the best feature and corresponding threshold that minimizes the error in predicting the target variable.

    Feature Selection: The decision stump evaluates each feature in the dataset to determine which one will be used for the split. It tests various potential thresholds within each feature's range. The threshold is a value that splits the data into two subsets: one where the feature's value is less than or equal to the threshold and one where it is greater.

    Threshold Determination: For each possible threshold, the decision stump calculates the mean target value of the data points that fall into the left and right regions (based on whether their feature value is less than or greater than the threshold). These mean values represent the predictions that the stump will make for each region.

    Error Minimization: The decision stump then calculates the Mean Squared Error (MSE) for each possible split. The MSE is computed by measuring the difference between the actual target values and the predicted mean values within each region, squared, and averaged over all data points. The stump selects the feature and threshold that result in the lowest MSE, effectively choosing the split that best separates the data in a way that minimizes prediction error.

    Prediction: Once the best feature and threshold are determined, the decision stump makes its predictions: it assigns the mean value of the target variable for the left region to all data points falling to the left of the threshold, and the mean value for the right region to all data points falling to the right.


#### Ramifications of This Implementation

While this simplified implementation is excellent for understanding the core mechanics of Gradient Boosting, it comes with several important ramifications and limitationws:

1. **Limited Model Complexity**:
    - Using a decision stump as the weak learner imposes significant limitations on the complexity of the model. Each weak learner can only capture very simple patterns in the data, which means the model may require many iterations to capture more complex relationships.
    - While this simplicity can help reduce overfitting, it also means that the model may underfit the data if not enough iterations are performed.

2. **Slow Learning Process**:
    - Due to the simplicity of decision stumps, each weak learner only makes small adjustments to the predictions. As a result, the learning process can be slow, requiring a large number of iterations to achieve satisfactory performance.
    - This slow learning process, moderated by the learning rate, is a trade-off that allows the model to gradually refine its predictions, reducing the risk of overfitting.

3. **Overfitting Considerations**:
    - The simplicity of decision stumps reduces the risk of overfitting in each individual learner, but the cumulative effect of many weak learners can still lead to overfitting, especially if the learning rate is too high or the number of iterations is too large.
    - In practice, methods such as cross-validation, early stopping, or regularization are commonly used to monitor and prevent overfitting.

4. **Interpretability**:
    - One of the advantages of using decision stumps is that the resulting model remains relatively interpretable. Each decision stump provides a simple rule (a single split) that can be easily understood and visualized.
    - However, as the number of iterations increases, the overall model becomes more complex, and the combined effect of all the weak learners may become harder to interpret. This is a trade-off between interpretability and model complexity.

5. **Scalability**:
    - While this simplified implementation is useful for educational purposes, it is not designed for scalability. In real-world applications, especially with large datasets or high-dimensional feature spaces, more efficient implementations such as XGBoost or LightGBM are necessary to handle the computational demands.
    - These advanced implementations address scalability through techniques like parallel processing, optimization algorithms, and efficient memory management, making them suitable for large-scale data analysis.
    
6. **MSE (or SSE) as the loss function**:
    - Using Mean Squared Error (MSE) as the loss function in Gradient Boosting has several potential drawbacks. MSE is highly sensitive to outliers because it squares the error terms, leading to disproportionately large impacts from extreme values. This can cause the model to overfit to outliers, compromising its generalization. Additionally, MSE can lead to unstable training due to large gradient magnitudes from high residuals, especially in early iterations. This can lead to oscillations or even divergence in the model's training, where instead of steadily improving, the model's performance fluctuates or degrades. MSE does not account for the distribution of errors or domain-specific cost considerations, which can be problematic in scenarios where different errors have different consequences. It also encourages increased model complexity to minimize all errors, potentially leading to overfitting, especially in noisy data. Therefore, MSE if quite a blunt instrument but can still work reasonably well; however, others should definitely be explored.

## Visualising the decision stump weak learner in a regression context

Similar to our AdaBoost example, a decision stump in a regression context is also just a simple decision tree with only one split. It works by selecting the best feature and threshold that splits the data into two regions, each with its own predicted value - **the predicted value is the mean of all the target values on each side of the threshold**. In the context of regression, the decision stump calculates the Mean Squared Error (MSE) for each possible split and chooses the one that minimizes this error. Again, this is where things differ from our previous usage of the decision stump - it then assigns the mean target value of the data points in each region as the prediction for that region.

```python
X = df_train[['total_vehicles_r_sum_3','total_vehicles_r_sum_6_slope_6']].values
y = df_train[['estimated_quarterly_percent_change']].values

plt.figure(figsize=(10, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', edgecolor='k')
plt.colorbar(label='Target Value')
plt.title('Original Dataset')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()

def plot_split(X, y, feature_index, threshold, mse, left_value, right_value, best=False):
    """Plot how the data is split by a threshold on a feature, including mean values for the left and right regions."""
    left_mask = X[:, feature_index] <= threshold
    right_mask = X[:, feature_index] > threshold

    plt.figure(figsize=(10, 6))
    plt.scatter(X[left_mask, 0], X[left_mask, 1], c=y[left_mask], cmap='viridis', edgecolor='k')#, label='Left Region')
    plt.scatter(X[right_mask, 0], X[right_mask, 1], c=y[right_mask], cmap='plasma', edgecolor='k')#, label='Right Region')
    
    # Plot the vertical line for splits on Feature 1
    if feature_index == 0:
        plt.axvline(x=threshold, color='r', linestyle='--', label=f'Threshold = {threshold:.2f}')
    
    # Plot the horizontal line for splits on Feature 2
    if feature_index == 1:
        plt.axhline(y=threshold, color='r', linestyle='--', label=f'Threshold = {threshold:.2f}')

    # Add mean values for the left and right regions
    plt.text(0.05, 0.95, f'Left Mean: {left_value:.2f}', ha='left', va='top', transform=plt.gca().transAxes, fontsize=12, color='blue')
    plt.text(0.95, 0.05, f'Right Mean: {right_value:.2f}', ha='right', va='bottom', transform=plt.gca().transAxes, fontsize=12, color='red')

    plt.colorbar(label='Target Value')
    plt.title(f'Split on Feature {feature_index + 1} at Threshold {threshold:.2f}\nMSE: {mse:.4f}' + (" (Best Split)" if best else ""), size=15)
    plt.xlabel('Feature 1', size=15)
    plt.ylabel('Feature 2', size=15)
    plt.legend(loc='center left', fontsize='small')
    plt.xticks(size=15)
    plt.yticks(size=15)
    plt.show()

# Visualize the process of finding the best split
best_mse = float('inf')
best_feature_index = None
best_threshold = None
best_left_value = None
best_right_value = None

# Iterate over each feature
for feature_index in range(X.shape[1]):
    thresholds = np.unique(X[:, feature_index])

    # Test each threshold as a potential split point
    vis_counter = 0
    for threshold in thresholds:
        left_mask = X[:, feature_index] <= threshold
        right_mask = X[:, feature_index] > threshold

        if np.sum(left_mask) == 0 or np.sum(right_mask) == 0:
            continue

        left_value = y[left_mask].mean()
        right_value = y[right_mask].mean()

        # Calculate mean squared error for this split
        mse = (
            np.mean((y[left_mask] - left_value) ** 2) * np.sum(left_mask) +
            np.mean((y[right_mask] - right_value) ** 2) * np.sum(right_mask)
        ) / len(y)

        # Visualize the split
        if vis_counter % 10 == 0:
            plot_split(X, y, feature_index, threshold, mse, left_value, right_value)
        vis_counter += 1

        # If this is the best split so far, store it
        if mse < best_mse:
            best_mse = mse
            best_feature_index = feature_index
            best_threshold = threshold
            best_left_value = left_value
            best_right_value = right_value

# Visualize the best split found with the regression prediction for both regions
plot_split(X, y, best_feature_index, best_threshold, best_mse, best_left_value, best_right_value, best=True)
```

## Define regression decision stump configured for MSE  

```python
def find_best_split(X, y):
    """Find the best feature and threshold to split the data."""
    X = np.array(X)  # Ensure X is a NumPy array
    n_samples, n_features = X.shape
    best_mse = float('inf')
    best_feature_index = None
    best_threshold = None
    best_left_value = None
    best_right_value = None

    # Iterate over each feature
    for feature_index in range(n_features):
        thresholds = np.unique(X[:, feature_index])

        # Test each threshold as a potential split point
        for threshold in thresholds:
            left_mask = X[:, feature_index] <= threshold
            right_mask = X[:, feature_index] > threshold

            if np.sum(left_mask) == 0 or np.sum(right_mask) == 0:
                continue

            left_value = y[left_mask].mean()
            right_value = y[right_mask].mean()

            # Calculate mean squared error for this split
            mse = (
                np.mean((y[left_mask] - left_value) ** 2) * np.sum(left_mask) +
                np.mean((y[right_mask] - right_value) ** 2) * np.sum(right_mask)
            ) / n_samples

            # If this is the best split so far, store it
            if mse < best_mse:
                best_mse = mse
                best_feature_index = feature_index
                best_threshold = threshold
                best_left_value = left_value
                best_right_value = right_value

    return best_feature_index, best_threshold, best_left_value, best_right_value, best_mse
```

```python
def train_weak_learner(X, residuals):
    """Train a decision stump (weak learner) on the residuals."""
    
    #========================
    #YOUR CODE HERE
    #========================
    X = np.array(X)  # Ensure X is a NumPy array
    n_samples, n_features = X.shape
    best_mse = float('inf')
    best_feature_index = None
    best_threshold = None
    best_left_value = None
    best_right_value = None

    # Iterate over each feature
    for feature_index in range(n_features):
        thresholds = np.unique(X[:, feature_index])

        # Test each threshold as a potential split point
        for threshold in thresholds:
            left_mask = X[:, feature_index] <= threshold
            right_mask = X[:, feature_index] > threshold

            if np.sum(left_mask) == 0 or np.sum(right_mask) == 0:
                continue

            left_value = residuals[left_mask].mean()
            right_value = residuals[right_mask].mean()

            # Calculate mean squared error for this split
            mse = (
                np.mean((residuals[left_mask] - left_value) ** 2) * np.sum(left_mask) +
                np.mean((residuals[right_mask] - right_value) ** 2) * np.sum(right_mask)
            ) / n_samples

            # If this is the best split so far, store it
            if mse < best_mse:
                best_mse = mse
                best_feature_index = feature_index
                best_threshold = threshold
                best_left_value = left_value
                best_right_value = right_value

    return best_feature_index, best_threshold, best_left_value, best_right_value, best_mse
```

## Make a prediction based on the entire ensemble 


**Exercise**: Implement the function that aggregates predictions from all weak learners (decision stumps) to make the final prediction. The function should start with the initial prediction, then iteratively add the contributions from each weak learner, scaled by the learning rate.

    initial_pred: The initial prediction value for all data points, often set as the mean of the target values.

    stumps: A list of tuples, where each tuple represents a weak learner (decision stump). Each tuple contains the feature index, threshold, left and right predicted values, and the best MSE (which can be ignored in this function).

    learning_rate: This scalar controls the influence of each weak learner’s prediction on the overall prediction.

    y_pred: This is the final prediction, starting with initial_pred and being adjusted by each stump.

```python
def predict(X, initial_pred, stumps, learning_rate):
    """Aggregate the predictions from all weak learners, applying the learning rate."""
    i_pred = np.full(len(X), initial_pred)
    for stump in stumps:
        i_pred = update_predictions(i_pred, stump[0], stump[1], stump[2], stump[3], X, learning_rate) #i managed to do this in one line because of list comprehension!
    
    return i_pred
```
