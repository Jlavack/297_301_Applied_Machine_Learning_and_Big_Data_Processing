SMOTEboost, AdaCost and SMOTECost as I call them. The cost adjustment penalty formula used is from Fan et al (1999):
   
   $$\large D_{t+1}(i) = \frac{D_{t}(i) e^{-\alpha_{t}y_{i}h_{t}\frac{x_{i}}{\beta(i)}}}{Z_t}$$
   
   - Additionally, I extended the implementation to provide each ensemble member in a dictionary in order to perform rank based pruning. And a step to automatically calibrate the probabilities using isotonic regression. The probabilities for this are dervied using this formula from (Schapire, Freund. 2014):
   
   $$\large P(y=1|x) = \frac{1}{1 + e^{-2 * F(x)}}$$
   
   $$\large F(x) = \sum^T_{t=1} \alpha_t H_t (x) $$

### 5. Implement an ensemble pruning/optimisation procedure that is executed after the model training : (5 marks)
   - I managed to implement pruning in the AdaBoost ensemble using two methods: iterative pruning to optimise a specific performance metric and rank pruning based on a series of novel measures. 
    * [Iterative refinement pruning](#1.5.1-Iterative-refinement-pruning)
    * [Rank pruning using novel ranking measures](#1.5.2-Rank-pruning-using-novel-ranking-measures)
   - With the synthetic dataset pruning can indeed improve the model with fewer members, however using a back pruning methods only ends up removing one or two ensemble members because by the last member diminishing returns is usually reached. To show a greedy pruning approach I genreated a graph showing the improvement in the score per addiitonal member for each pruning method. 
   - Iterative pruning shows that, when comparing AdaBoost, SMOTEBoost and ADACost up to a maximum of 20 classifiers (20 chosen both for computioant time and because we hit diminished returns around 20), we see peak performance on any given measure earlier than 20 members. So that we can take the stance that additional members add no benefit to the ensemble. For example, error minimises around 7 members (not in the graphs both AdaBoost and SMOTEBoost are identical in their pruned performance given the SMOTEBoost parameters of 2 new samples from 20 neigbours with weight 0.0001). 
   - Accuracy, F1, precioins and MCC peak between 7 and 13 (two maxima). While ROC AUC continues to rise for SMOTE and AdaBoost, but hits a maximum for ADACost at 7. So, in general, we see that 7 members is ideal for most metrics. Additional members add no benefit to the ensemble. Unless one is optiminising for ROC AUC. 
   - Rank pruning follows a different approach. It ranks each ensemble member by their contribution to the ensemble, then takes the best 'expert learners'. The idea for this came from the Cohen Kappa metric. I thought that each ensemble memeber is a bit like someone giving their opinion about something and if I could remove the least useful voters then the model would be better. As it turns out, someone else also had this idea and wrote a paper on it. Then, I found other methods to do the same with different measures, such as margin density maximisation which is specific to AdaBoost. 
   - these measures are MDM, weighted MDM, Independent contribution (IC), Class error, Begative AUC, Erro ambiguity, Cohen's Kappa and reference vector accuracy. 
   - So I implemented each of these measures and tried rank pruning on a vanilla ensemble with 30 iterations to see if any meta measure aided in optimising for a particular performance metric. 
   - For all measures the improvement happens with 25 to 30 ensemble memebers. Not vastly smaller like with iterative pruning, unless one wants to optimise for the ROC AUC, which can max out with 15 members on the class error measure. 
   - Basically, each measure is an additive one on some level. And the optimisation looks to find the overall ensemble performance with the top x ensemble members. Given the additive nature, the overall improvement is low until we hit high numbers of ensembles realtive ot the datset size because when ensemble members are few each new member adds something useful. Here again we find diminsihing returns. As the number of ensemble memebers increases we capture more variability. As long as each new member adds new variability, there isn't a reason to remive them. As soon as adding a new one (the nth ranked member) changes this, we don't remove them. 
   - In theory, one could apply ank pruning first and then, with the newly ranked ensemble iteratively prune the ensemble back. However, I didn't try this, because the comparative results of an unpruned model and the two pruning methods indicated this was a bad idea. While with either approach one can get high scores such as accuracy or preciison with fewer ensemble members, they aren't better overall. Which is because they optimiase by definition on one score. For example, with 26 ranked pruned members or 7 iteratively pruned members we can get a precision of 0.98 each compared to 0.96 with an unpruned model. While the MCC, accuracy and F1 scores all decrease, as does the ROC AUC. Of interesting note is the pruned models become very good at recalling the negative class at 0.99, but less (0.5 ish) good with the positive class. For each methods the reason is likely different. Rank pruning finds expert learners and if these learners are better on one class, this is reflected in this way. Iterative pruning cuts out learners, and so stops the model getting better on the other class. This occurs, I expect, more prominently with imbalanced classes.

```python
#helper functions that will be used often
def binary(xi):
    if xi == -1: 
        return 0
    else:
        return 1

def convert(xi):
    if xi == 0:
        return 1
    else:
        return -1

def misclassify(x, percentage):
    y=x.copy()
    num_elements = int(len(y) * percentage / 100)
    indices = np.random.choice(len(y), num_elements, replace=False)
    y[indices] *= -1
    return y

def sigmoid(x):
    return 1/(1+np.exp(-x))

def euclidean_dist(x1, x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))

#used for SMOTE
def find_neighbours(data, row, k): 
    distances = []

    for i in data:
        dist = euclidean_dist(row, i)
        distances.append(dist)
        
    distances = np.array(distances)
    
    idx = np.argsort(distances) 
    distances = np.sort(distances)

    neighbours = []
    nbr_idx = []
    for i in range(k):
        neighbours.append(distances[i])
        nbr_idx.append(idx[i])
        
    neighbours = np.array(neighbours)
    nbr_idx = np.array(nbr_idx)
        
    return neighbours, nbr_idx

def pred_probs(alphas): #takes the sum of the alphas and does quick and easy prediction of probabilities. Avoids complex calcs. 
    prob_pred_pos = sigmoid(-2 * (-1 * alphas))
    prob_pred_neg = 1 - prob_pred_pos
    
    return prob_pred_pos, prob_pred_neg

def deep_get(dictionary, *keys, default=None): #getting things out of dictionaries
    for key in keys:
        if isinstance(dictionary, dict):
            dictionary = dictionary.get(key, default)
        else:
            return default
    return dictionary
```

## ADABoost base functions

```python
def calculate_decision_stump(data, weights, labels):
    """
    Calculate the best decision stump for a given feature.
    
    Args:
    data : Dataset containing just as single feature
    weights: Sample weights
    labels: Sample class labels
    
    Returns:
    tuple: (best_threshold, best_direction, min_error)
    """
    data = pd.DataFrame(data)
    min_error = float('inf')
    best_thresh = 0
    best_dir = 0
    
    # Calculate thresholds
    feature_min = data.min().iloc[0]
    feature_max = data.max().iloc[0]
    #print("Min: ", feature_min, "max: ", feature_max)
    interval = (feature_max - feature_min) / 100.0
    #print("INTERVAL: ", interval)
    thresholds = pd.interval_range(start=feature_min-interval*2, end=feature_max+interval*2, freq=interval)  #optimise thresholds?
    #print(thresholds)
    
    for direction in [1, -1]:
        for thresh in thresholds:
            if direction == 1:
                predictions = (data >= thresh.left).astype(int) * 2 - 1 #this transforms the values from {0, 1} to {-1, 1}. Specifically, 1 remains 1, and 0 becomes -1.
            else:
                predictions = (data < thresh.left).astype(int) * 2 - 1 #this transforms the values from {0, 1} to {-1, 1}. Specifically, 1 remains 1, and 0 becomes -1.
           # print("PREDICTIONS ", [thresh], ": ", predictions, """""")
            
            # Calculate misclassifications    
            #print(len(predictions), " : ", labels.shape)
            misclassifications = (predictions.iloc[:,0] != labels).astype(int)
            #print("MISSCLAS: ", misclassifications, """""")
            
            # Calculate weighted error
            weighted_error = (misclassifications * weights).sum()
            
            #print("WEIGHTED_ERROR: ", weighted_error, """""")
            if weighted_error < min_error:     
                min_error = weighted_error
                best_thresh = thresh.left
                best_dir = direction    
    
    return best_thresh, best_dir, min_error
```

```python
def calculate_alpha(weighted_error):    
        return 0.5 *np.log((1-weighted_error)/ weighted_error) if weighted_error != 0 else 0

def classify_dataset_against_weak_classifier(x, thresh, direction):
    classification = np.zeros(len(x))
    
    #classifiy all samples based on the last feature
    #get actual classification
    for i in range(len(x)):   
        if direction == -1:
            if x[i] < thresh:
                classification[i] = 1
            else: 
                classification[i] = -1
        else:
            if x[i] > thresh:
                    classification[i] = 1
            else: 
                classification[i] = -1  
                
    return classification 

def update_weights(weight, alpha, classification, label):
    for i in range(len(weight)):
        weight[i] =  weight[i] * np.exp(-1.0 * alpha.item() * classification[i] * label[i])
    return weight

def normalise_weights(weight):
    weight = weight/np.sum(weight)
    return weight 

def classify_sample(xi, boost_classif):
    boost_classif = boost_classif.values
    classification_sum = np.float64(0)
    
    for thresh, feat, sign, alpha, error in boost_classif:       
        ht = np.sign(xi[int(feat)] - thresh) * sign
        classification_sum += alpha * ht
        
    if classification_sum >= 0:
        return 1
    else:
        return -1
    
def predict_Adaboost(eval_data, classifier_df, fit_df):
    predictions = []
    
    for i in range(len(eval_data)):
        x = classify_sample(eval_data[i], classifier_df)
        predictions.append(x)
        
    return predictions
```

```python
def sum_classifier_votes_for_each_sample(dataset, df, classifier_df):
    for i in range(len(dataset)):
        classification_sum = np.float64(0)
        neg_votes = np.float64(0)
        pos_votes = np.float64(0)
        for idx, thresh, feat, sign, alpha, error in classifier_df.itertuples():
             
            ht = np.sign(dataset[i, int(feat)] - thresh) * sign
            classification_sum += alpha * ht
            if ht < 0:
                neg_votes += alpha * ht
            else:
                pos_votes += alpha * ht

        df['sum_alpha'].iloc[i] = classification_sum
        df['pos_votes'].iloc[i] = pos_votes
        df['neg_votes'].iloc[i] = neg_votes
    
    return df

def margin_calculation(sign, pos, neg, tot_votes):      
    if np.sign(sign) < 0:
        return np.abs(neg) / tot_votes, -1
    else:
        return pos / tot_votes, 1 
    
def sign_of_margin(margin, classification, true_class_label):  
    return margin * classification * true_class_label
```

```python
def predict_ensemble(ensemble, data, label): #fits the data and creates dataframe. If using train test split, use the model output and call this to get the results
    df = pd.DataFrame(data)
    df['sum_alpha'] = 0 
    df['pos_votes'] = 0 
    df['neg_votes'] = 0 

    df = sum_classifier_votes_for_each_sample(data, df, ensemble)
    total_alpha_votes = np.sum(ensemble.alpha)
    df['classification'] = 0
    df['margin'] = 0
    df['total_alpha_votes'] = total_alpha_votes
    result = df[['sum_alpha','pos_votes','neg_votes','total_alpha_votes']].apply(lambda data: margin_calculation(*data), axis=1)
    df['margin'] = result.apply(lambda data: data[0])
    df['classification'] = result.apply(lambda data: data[1])
    df['true_class_label'] = label
    df['sign_of_margin'] = df[['margin', 'classification', 'true_class_label']].apply(lambda data: sign_of_margin(*data), axis=1)
    df['probabilities'] = df.apply( lambda x: 1.0 -x.margin if x.classification == -1 else x.margin, axis = 1)   
    df['pred_prob_pos'], df['pred_prob_neg'] = pred_probs(df['sum_alpha'])
    df['cal_prob'] = isotonic_calibration(df['true_class_label'], df['pred_prob_pos'])
    
    return df
```

## ADABoost function
[Back to top](#Index)  |  [Back to AdaBoost implementation](#Task-1:-ADABOOST-IMPLEMENTATION)

```python
#incroporate pruning
#make it kfold validate naturally
def Adaboost(T, data, label, SMOTE_config, COST_config, members):
    SMOTE_flag = False
    COST_flag = False
        
    N=data.shape[0]
    dim = data.shape[1]
    h = np.zeros([T, 3], dtype=np.float64)
    alpha = np.zeros(T, dtype=np.float64)
    weight = ((pd.Series([1]*N))/N).values
    err = np.ones(T, dtype=np.float64) * np.inf 

    ensemble_list = [] #to hold each ensemble member
    
    if SMOTE_config is not None:
        SMOTE_n = SMOTE_config['new samples']
        SMOTE_k = SMOTE_config['neighbours']
        SMOTE_w = SMOTE_config['new weights']
        SMOTE_flag = SMOTE_config['SMOTE'] #could just set to True, but this allows for setting it off if hyper tuning
    
    if COST_config is not None:
        misclassification_costs = np.zeros(N)
        cost = np.full(N, COST_config['cost_penalty'])
        COST_flag = COST_config['COST']

    #0 - for each boosting round
    for t in range(T):
        #1 - iterate through every feature  
        for feature in range(dim):
            weighted_error = np.float64(0)

            #========================
            #1 - GENERATE A DECISION STUMP FOR A FEATURE
            #========================
            threshold, sign, weighted_error = calculate_decision_stump(data[:,feature], weight, label)
            #returns best_thresh, best_dir, min_error

            #========================
            #2 - KEEP TRACK OF THE FEATURE WITH THE LOWEST WEIGHTED ERROR
            #========================   

            if weighted_error < err[t] : 
                err[t] = weighted_error         
                h[t] = [threshold, feature, sign]

        #========================
        #3 - CLASSIFY ALL SAMPLES BASED ON THE SELECTED FEATURE FOR BOOSTING ROUND t
        #========================
        classification = classify_dataset_against_weak_classifier(data[:,h[t][1].astype(int)], h[t][0], h[t][2])
        #returns classification as array

        #========================
        #COST 1 - CALCULATE MISCLASSIFICATION COSTS
        #========================
        #update the weighted error based on cost formula
        if COST_flag == True:
            err[t], misclassification_costs = misclassification_cost(cost, weight, classification, label, misclassification_costs)
        
        #========================
        #4 - CALCULATE ALPHA FOR BOOSTING ROUND t (if Cost is on, err[t] will be different)
        #========================            
        alpha[t] = calculate_alpha(err[t])
        #returns alpha as a value
        
        #========================
        #5 - UPDATE WEIGHTS BASED ON THE CORRECTNESS OF THE CLASSIFICATION
        #========================
        if COST_flag == False:
            weight = update_weights(weight, alpha[t], classification, label)
            #returns weight as an array
        if COST_flag == True:
            weight = ADACost_update(weight, cost, alpha[t], classification, label)

        #========================
        #6 - NORMALISE REASSIGNED WEIGHTS
        #========================  
        weight = normalise_weights(weight)
        #returns weight as array 

        #put the current model into a list for pruning
        if members == True:
            cdf = pd.DataFrame(pd.DataFrame([[h[t][0],h[t][1],h[t][2],alpha[t],err[t]]], columns= ["threshold", "feature", "direction","alpha", "error"]))
            cdf = predict_ensemble(cdf, data, label)
            ensemble_list.append(cdf)
        
        #========================
        #7 - SMOTE: if the n is set too high, the SMOTE algo will swap the minority class automatically
        #========================
        if SMOTE_flag == True:
            SMOTE_data, SMOTE_label = SMOTE(data, label, SMOTE_n, SMOTE_k)
            data = np.append(data, SMOTE_data, axis=0)
            label = np.append(label, SMOTE_label)
            weight = SMOTE_weights(weight, SMOTE_n, SMOTE_w)
            if COST_flag == True:
                misclassification_costs = np.pad(misclassification_costs, pad_width = (0, SMOTE_n), constant_values = 0)
                cost = np.pad(cost, pad_width = (0, SMOTE_n), constant_values = COST_config['cost_penalty'])
                
        #--------------------------------------------
        #BOOSTING ALGORITHM DONE, NOW CREATE OUTPUT DATAFRAMES
        #--------------------------------------------
    classifier_df = pd.DataFrame(h, columns = ["threshold", "feature", "direction"])
    classifier_df["alpha"] = alpha
    classifier_df["error"] = err
    training_set_df = predict_ensemble(classifier_df, data, label)
        
    print("Classification of ", T, " iterations complete")
    
    if members is True: return classifier_df, training_set_df, ensemble_list
    else: return classifier_df, training_set_df
```

```python
def hypothesis(data, theta, bias): 
    return sigmoid(np.dot(theta, data.T)+bias) #need to output per data row

def cost(hypothesis, label): #error
    return (-label * np.log(hypothesis) - (1 - label) * np.log(1 - hypothesis)).mean()

def train_logistic(data, label, iterations, learning_rate, weight):
    #print(data.shape)
    theta = 0  #np.zeros(data.shape[1])
    bias = 0 #initialise 0 bias
    data = np.array(data)

    for i in range(iterations):
        cost_list = []
        H = hypothesis(data, theta, bias)
        #gradient_descent
        cost = (-label * np.log(H) - (1 - label) * np.log(1 - H)).mean()
        cost_list.append(cost)
        gradient_t = np.dot(data, (H - label)*weight) / len(label)
        #gradient_b = np.sum(H - label) / len(label)
        theta -= learning_rate * gradient_t
        theta = theta.sum()
    return theta, cost_list

def predict_logistic(data, theta, bias, threshold):
    predictions = hypothesis(data, theta, bias)
    H = predictions.copy()
    predictions = np.where(predictions >= 0.5, 1, -1)
            
    return predictions, H

def logistic_class(data, label, iterations, learning_rate, threshold, weight):
    theta, cost_list = train_logistic(data, label, iterations, learning_rate, weight)
    bias = 0
    result, H = predict_logistic(data, theta, bias, threshold)

    return result, cost_list, H, theta#, bias, H

def Siggy_stump(data, weights, label, T):
    data = pd.DataFrame(data)
    min_error = float('inf')
    best_thresh = 0
    best_dir = 0
            
    # Calculate random thresholds
    feature_min = data.min().iloc[0]
    feature_max = data.max().iloc[0]
    interval = (feature_max - feature_min) / 100.0
    random_starts = np.random.uniform(feature_min, feature_max, size=1)
    
    thresholds = pd.IntervalIndex.from_arrays(random_starts, random_starts + interval, closed='left')
    d = np.array(data[0])
    df = pd.DataFrame({'value': d, 'label': label, 'weights':weights})
#     print(df['value'])
#     print(df)
    for direction in [1, -1]:
        for threshold in thresholds:
            if direction == 1:
                threshold = threshold.left
                df_clipped = df[(df['value'] >= threshold) & (df['value'] <= feature_max)]
#                 print(len(df_clipped))
                predictions, siggy_cost, H, theta = logistic_class(np.array(df_clipped['value']), np.array(df_clipped['label']), 5, 0.1, threshold, np.array(df_clipped['weights']))
                misclassifications = (predictions == np.array(df_clipped['label'])).astype(int)
                weighted_error = (misclassifications * np.array(df_clipped['weights'])).sum()

            else:
                threshold = threshold.right
                df_clipped = df[(df['value'] >= feature_min) & (df['value'] <= threshold)]
                predictions, siggy_cost, H, theta = logistic_class(np.array(df_clipped['value']), np.array(df_clipped['label']), 5, 0.1, threshold, np.array(df_clipped['weights']))
#                 print("PREDS: ", predictions)
                misclassifications = (predictions == np.array(df_clipped['label'])).astype(int)
                # Calculate weighted error
                weighted_error = (misclassifications * np.array(df_clipped['weights'])).sum()

            if weighted_error < min_error:     
                min_error = weighted_error
                best_cost = siggy_cost
                best_theta = theta   
                best_thres = threshold
                best_dir = direction

    return best_cost, best_theta, min_error, best_thres, best_dir
    
```

```python
#SMOTE CODE
#define function for SMOTE resampling to make more blue. pass in minoirty sample only.
#then add n many minorty labels onto the label dataset

def SMOTE(data, label, n, k):
    #random.seed(21)
    #n = samples to create
    #k = number of neighbours to generate samples from
    
    #identify and prune data for minoirty class only
    minority_class = max(Counter(label).items(), key=lambda x: x) #not sure why, but min and max are inverted for this imported function
    minority_class[0]
    
    data = data[(np.nonzero(label == minority_class[0]))]
    
    shape = (n, data.shape[1])
    samples = np.empty(shape = shape)
    idx = 0
    
    for idx in range(samples.shape[0]):
        max_idx = data.shape[0]
        A_idx = np.random.randint(low = 0, high = max_idx)
        A_point = data[A_idx]
        distances, nbr_idx = find_neighbours(data, A_point, k+1 )
        #nbr_arr = nbr_idx[0]

        if A_idx in range(nbr_idx.shape[0]):
            condition = np.where(nbr_idx == A_idx)
            nbr_arr = np.delete(nbr_idx, condition)
            
        len_nbr_arr = nbr_idx.shape[0]
        if len_nbr_arr > 0:
            b_idx = np.random.randint(low=0, high = len_nbr_arr)
        else:
            b_idx = 0
        
        B_point = data[nbr_idx[b_idx]]
        m = (B_point[1] - A_point[1]) / (B_point[0] - A_point[0])
        
        high_point = A_point[0] if A_point[0] > B_point[0] else B_point[0]
        low_point = A_point[0] if A_point[0] < B_point[0] else B_point[0]
        
        if m==m: 
            random_x = np.random.uniform(low=low_point, high=high_point)
            random_y = m * (random_x-A_point[0]) + A_point[1]
        else:
            random_x = B_point[0]
            random_y = B_point[1]
        
        samples[idx] = B_point#(random_x, random_y) #turn off x,y output for larger feature inputs. 
        
    labels = np.full(n, minority_class[0]) # make the new labels for appending
        
    return samples, labels

#this function adds new samples to the weights arracy at w and renormalises the array at the new length. 
def SMOTE_weights(weights, n, w):
    SMOTE_weights = np.full(n, w)
    new_weights = np.append(weights, SMOTE_weights)
    new_weights = normalise_weights(new_weights)

    return new_weights
```

```python
#COST CODE
def misclassification_cost(cost, weight, classification, label, misclassification_costs):
    
    for i in range(len(misclassification_costs)):
        if label[i] != classification[i]:
            misclassification_costs[i] = cost[i]
        else:
            misclassification_costs[i] = 0
    
    weighted_error = np.sum(weight * misclassification_costs)
    
    if weighted_error == 0:
        weighted_error - 0.0001 #avoid division by zero
      
    return weighted_error, misclassification_costs

def ADACost_update(weight, cost, alpha, classification, label):
    beta = np.ones(label.shape[0])
    
    for i in range(label.shape[0]):
        if label[i] == classification[i]:
            beta[i] = 1.0 / (1.0 + cost[i])
        else:
            beta[i] = 1.0 + cost[i]
    
    exp = -1*alpha * label * classification * beta
    #exp = -1*alpha * label * classification / beta  #another method which works
    weight *= np.exp(exp) #update weights
    #weight *= beta * np.exp(exp) #another update method which seems to be really good, so I am suspicious of overfitting
    weight /= np.sum(weight)  #normalise the weights
    
    return weight
```

```python
def k_fold_split(data, label, folds, seed):
    np.random.seed(seed)  
    k_fold_dict = {}
    indices = np.arange(len(data))
    np.random.shuffle(indices)
    fold_size = len(data) // folds
    remainder = len(data) % folds

    start_idx = 0
    for fold in range(folds):
        fold_dict = {fold: {}}
        current_fold_size = fold_size + 1 if fold < remainder else fold_size
        #test_start = fold * fold_size
        end_idx = start_idx + current_fold_size #((fold + 1) * fold_size)
        
        test_indices = indices[start_idx:end_idx]
        train_indices = np.concatenate([indices[:start_idx], indices[end_idx:]])     
        
        #Split data into training and test sets    
        fold_dict[fold]["X_train"] = data[train_indices]
        fold_dict[fold]['X_test'] = data[test_indices]
        fold_dict[fold]['y_train'] = label[train_indices]
        fold_dict[fold]['y_test'] = label[test_indices]
        
        k_fold_dict.update(fold_dict)
        start_idx = end_idx
                
    return k_fold_dict #fold_size, indices

#CV Kfold process
#get the training data and training label for fold f and train the model
#Calculate the metrics with model_performance
#add each model to a dictionary with its ensemble, predictions and ensemble members
#add the results to a dictionary fith {fold {model_performance}}

def cross_validation(dictionary, T, SMOTE_config, COST_config, members):
    CV_dict = {}

    #part here to unpack a dictionarty of commands for the ADAboost call

    for fold in dictionary:
        #print(f"Validating fold {fold}")
        nested_dict = {fold: {}}
        X_train = dictionary[fold]['X_train']
        X_test = dictionary[fold]['X_test']
        y_train = dictionary[fold]['y_train']
        y_test = dictionary[fold]['y_test']
        
        if members is True:
            model, fit, members = Adaboost(T, X_train, y_train, SMOTE_config, COST_config, members) #train the model
        else: model, fit = Adaboost(T, X_train, y_train, SMOTE_config, COST_config, False)
        #next step will also give a calibrated probability
        predictions = predict_ensemble(model, X_test, y_test) #result classifications are the predictions and true class lable the truth
        fitted_scores = model_performance(fit)
        predicted_scores = model_performance(predictions)

        #add to dictionary
        nested_dict[fold]['model'] = model
        nested_dict[fold]['fit'] = fit
        if members is True: nested_dict[fold]['members'] = members
        nested_dict[fold]['predictions'] = predictions
        nested_dict[fold]['fitted_metrics'] = fitted_scores
        nested_dict[fold]['predicted_metrics'] = predicted_scores
        CV_dict.update(nested_dict)
    return CV_dict

def get_from_cv(dictionary, key1, key2): #key1 could be members or metrics, key2 is the member or the metric
    retrieved = []
    for fold in dictionary:
        if key2 is not None:
            x = dictionary[fold][key1][key2]
            retrieved.append(x)
        else: 
            x = dictionary[fold][key1]
            retrieved.append(x)
    return retrieved
```

### Grid search hyper tuning functions

```python
#Hyper tuning grid searhc
def get_grid_list(key_list, value_list): #gets all possible combos in dictionary format so we can call AdaBoost
    combinations = list(itertools.product(*value_list))
    grid_list = [] 
    for i in range(len(combinations)):
        grid_list.append(dict(zip(key_list, combinations[i])))
    return grid_list

def get_param_grid(params): #get a list of lists where each list is all combos in the params dictionary. 
    params_list = []
    for key in params.keys():
        key_list = params[key].keys()
        value_list = params[key].values()
        combo = get_grid_list(key_list, value_list)
        params_list.append(combo)
        
    param_grid = list(itertools.product(*params_list))
    return param_grid
        

def grid_search(T, X_train, y_train, params, folds, seed, members):
    best_score = 0
    best_params = None
    
    #get k fold splits fof CV
    kf = k_fold_split(X_train, y_train, folds, seed)
    
    #make a grid per SMOTE or COST
    param_grid = get_param_grid(params)
    
    #de-duplicate and add a (None, None) value in... this will save a bit of time in tuning
    param_grid = [(None, tuple2) if any(value is False for value in tuple1.values()) else (tuple1, tuple2)
    for tuple1, tuple2 in param_grid] #first side of tuple
    param_grid = [(tuple1, None) if any(value is False for value in tuple2.values()) else (tuple1, tuple2)
    for tuple1, tuple2 in param_grid] #other side of tuple
    param_grid = [sub for sub in param_grid if not all(ele is None for ele in sub)] #remove duplicates
    param_grid.insert(0, (None, None)) #add the standard config back

    #hypertune. Every train test split on every parameter set... long process so best ot have an idea of what you want first. 
    CV_results = {}
    for i in range(len(param_grid)):
        nested_dict = {f'tune round {i}':{}}
        print(f"TUNING ROUNG {i+1} of {len(param_grid)}")
        CV_dict = cross_validation(kf, T, param_grid[i][0], param_grid[i][1], members)
        nested_dict[f'tune round {i}'] = CV_dict
        CV_results.update(nested_dict)
        #may need to add in here call to get metrics and index of model with params
        
    return CV_results, param_grid #param_grid will provide indexes to find the best set of params
```

```python
#iterative refinement pruning
def eval_ensemble(ensemble, data, label, key):
    train_df = predict_ensemble(ensemble, data, label)
    
    #calculate metrics
    metrics_dict = model_performance(train_df)
    
    return metrics_dict[key]

def prune(ensemble, data, label, key):
    baseline = eval_ensemble(ensemble, data, label, key)
    best_score, removed = baseline, None
    print("Baseline: ", baseline)
    
    #iterate through the ensemble and find a better score, if there is one
    for i, row in ensemble.iterrows():
        dup = ensemble.copy()
        dup = dup.drop(i)
        #print(dup)
        result = eval_ensemble(dup, data, label, key)
        imp = False
        
        if key == 'Error' or key == 'Log Loss' or key == "Brier Score Uncal" or key == "Brier Score Cal":
            if result < best_score: 
                imp = True
                best_score = result
                removed = i
            print("droped: ", i, " : result: ", result, " : improvement: ", imp, " : best score: ", best_score)

        else:
            if result > best_score: 
                imp = True
                best_score = result
                removed = i
            #print("droped: ", i, " : result: ", result, " : improvement: ", imp, " : best score: ", best_score)

    return best_score, removed

def prune_ensemble(ensemble, data, label, key):
    best_score = 0.0
    #print(f'Pruning using {key} metric')
    while True:  #prune until no further improvement
        score, removed = prune(ensemble, data, label, key)
        
        if removed is None:
            #print("No further improvement")
            break
        
        best_score = score
        ensemble = ensemble.drop(removed)
        #print('>%.3f (removed: %s)' % (score, removed))
        
    return best_score, ensemble

def get_iterative_prune_results(n, models_list, metric_list, model_dict, X_test, y_test):
    iter_refine_prune_metrics_dict = {}

    for mod in models_list:
        nested_dict = {mod: {}}

        for m in metrics_list: #I bet I can wrap this inside the later code for a more efficient funciton. figure this out
            metric_list = []

            for i in range(n):
                #print(f'ensemble growth: {mod}, {m}, {i+1}:')
                trained, training_ensemble = grow_ensemble(model_dict[mod], X_test, y_test, i+1)
                partial_ensemble_performance = model_performance(trained)
                metric_list.append(partial_ensemble_performance[m])
                nested_dict[mod][m] = metric_list

        iter_refine_prune_metrics_dict.update(nested_dict)
    return iter_refine_prune_metrics_dict
```

```python
#novel metrics for rank pruning of ensemble
def get_votes(data, dictionary):
    #have to get the ength of last dataset to account for SMOTE resmapling which otherwise puts everything out of bounds.
    neg = np.zeros(len(dictionary[-1]), dtype=int)
    pos = np.zeros(len(dictionary[-1]), dtype=int)

    for i, df in enumerate(dictionary):
        for j, row in dictionary[i].iterrows():
            #print(test5_list[i]['classification'][j])
            #print(df_test["1"][j])

            if dictionary[i]['classification'][j] == 1:
                pos[j] +=  1  #df["pos"][j] += 1
                #print(j)

            if dictionary[i]['classification'][j] == -1:
                neg[j] += 1 #df["neg"][j] += 1
                
    Votes = np.column_stack((pos, neg))  #df[["pos", "neg"]].to_numpy()
    return Votes

def get_votes_weighted(data, dictionary, truth):
    l = len(dictionary[-1])
    Votes = np.zeros((l, (np.unique(truth)).shape[0])) #array with sample n rows and one column per class
    for i, df in enumerate(dictionary):
        for j, row in dictionary[i].iterrows():
            Votes[j, 0] += dictionary[i]['pos_votes'][j]
            Votes[j, 1] += dictionary[i]['neg_votes'][j]
    return Votes

def get_ensemble_probs(dictionary): #Pass in dictionary i from iterative function call. necessary for Kappa calc, returns array with dimensions rows, classes, classifiers
    l = len(dictionary[-1])

    array_list = []
    #print(len(dictionary))
    for i in range(len(dictionary)):
        arr = np.column_stack((dictionary[i]['pred_prob_pos'], dictionary[i]['pred_prob_neg']))
        arr = np.pad(arr, ((0, l - len(arr)), (0, 0)), mode='constant', constant_values=0)
        array_list.append(arr)
        #print(f"len arr {i}", len(arr))

    ensemble_probs = np.stack(array_list, axis=0)
    #ensemble_probs[0:][0:][0] accesses entire first 2D array
    return ensemble_probs

def calc_MDM(Votes, predictions, truth, alpha): #calc maximum margin density, maybe need to pass an index to access right list
    n = predictions.shape[0] #try this with the margin and votes columns of the classifiers, see if different result
    MDM = 0
    #trim votes to length of truth
    l = len(truth)
    Votes = Votes[:l]
    #print("Len Votes, preds, Truth: ", len(Votes), len(predictions) ,len(truth))
     
    for j in range(n):
        if(predictions[j] == truth[j]):
            if (predictions[j] == np.argmax(Votes[j,:])): #special case where margin for label with majority votes
                sortedArray = np.sort(np.copy(Votes[j,:])) #calculate margin for second highest votes

                if (sortedArray[-2] == np.max(Votes[j,:])): #check if first and second max votes are equal, (margin =0)
                    margin = (Votes[j, binary(truth[j].astype(int))] - (sortedArray[-2] - 1)) / n
                else:
                    margin = (Votes[j, binary(truth[j].astype(int))] - (sortedArray[-2])) / n
            else: #usual case
                margin = Votes[j, binary(truth[j].astype(int))] - np.max(Votes[j,:]) / n

            if margin == 0: #case where margin is 0
                margin = 0.01
            fm = np.log(abs(margin))
            fd = np.log(Votes[j, binary(truth[j].astype(int))])
            if math.isinf(fd): #avoid infinity situations
                fd = 0
            fd = fd if math.isnan(fd) is False else 0 #avoid Nan situations
            MDM = MDM + (alpha * fm) + ((1-alpha)*fd)
            
            #print("margin, fm, fd, MDM: ", margin, fm, fd, MDM)
    
    return -1.0 * MDM

def weighted_MDM(Votes, predictions, truth, alpha):
    n = predictions.shape[0]
    MDM = 0.0
    #trim votes to length of truth
    l = len(truth)
    Votes = Votes[:l]
        
    for j in range(n):
         if(predictions[j] == truth[j]):
            if (predictions[j] == np.argmax(Votes[j,:])): #special case where margin for label with majority votes
                sortedArray = np.sort(np.copy(Votes[j,:])) #calculate margin for second highest votes

                if (sortedArray[-2] == np.max(Votes[j,:])): #check if first and second max votes are equal, (margin =0)
                    margin = (Votes[j, binary(truth[j].astype(int))] - (sortedArray[-2] - 1)) / n
                else:
                    margin = (Votes[j, binary(truth[j].astype(int))] - (sortedArray[-2])) / n
            else: #usual case
                margin = Votes[j, binary(truth[j].astype(int))] - np.max(Votes[j,:]) / n

            if margin == 0: #case where margin is 0
                margin = 0.01
            fm = np.log(abs(margin))
            fd = np.log(Votes[j, binary(truth[j].astype(int))])
            if math.isinf(fd): 
                fd = 0 #aviid infinity situations
            fd = fd if math.isnan(fd) is False else 0 #avoid Nan situations
            MDM = MDM + (alpha * fm) + ((1-alpha)*fd)
   
    return MDM

def calc_IC(Votes, predictions, truth): #calculate individual contributiosn of each sample to margin
    n = predictions.shape[0]
    IC = 0
    l = len(truth)
    Votes = Votes[:l]
    
    for i in range(n):
        if (predictions[i] == truth[i]):
            if (predictions[i] != np.argmax(Votes[i,:])): #case 1, minority group
                IC = IC + (2*(np.max(Votes[i,:])) - Votes[i, predictions[i]])
            else: #case 2, majority group
                sortedArray = np.sort(np.copy(Votes[i,:]))
                IC = IC + (sortedArray[-2])
        else: #case 3, wrong prediction
            IC = IC + (Votes[i, truth[i].astype(int)] - Votes[i, predictions[i]] - np.max(Votes[i,:]))
    
    return -1 * IC

def cal_classifier_error(predictions, truth): #may need an index to access right df in dictionaty
    return (predictions != truth).mean()

def negative_AUC(predictions, truth):
    #if predictions.shape[1] == 2: #extend to handle multiple classes by passing in probabilities for each class
    AUC = metrics.roc_auc_score(truth, predictions)
    return -1.0 * AUC

def error_ambiguity(i, ensemble_probs, truth):
    l = len(truth)
    xi = ensemble_probs[i,:,:] #get the predictions for the ith ensemble member
    xi =  xi[:l]
    #trim ensemble_probs down fro SMOTE
    
    x_hat = ensemble_probs.mean(axis = 0) #get the average probs for each sample class across all ensemble members
    x_hat =  x_hat[:l]

    sqrd_errors = (xi - x_hat)**2
    C = xi.shape[1] #number of classes
    
    A = 1.0 / C**2 * np.exp(-1.0 / C * xi)
    B = 1.0 / C**2 * (1.0 / (C-1))**2 * np.exp(1.0 / C * 1.0 / (C-1) * xi)
    mask = np.zeros(A.shape)
    np.put_along_axis(mask, truth[:, None].astype(int), 1.0, axis = 1)
    return (mask * A + (1.0 - mask) * B).sum() + sqrd_errors.sum()

def kappa_statistic(i, ensemble_probs, predictions, truth): #calculate Cohnen kapp stat for each classifier with respect to the entire ensemble
    scores = []
    l = len(truth)
    #print("Length truth: ", l)
    for j, jprobs in enumerate(ensemble_probs):
        if j != i: #if I'm not looking at the ith array in ensemble_array
            predK = jprobs.argmax(axis=1) #conver to predictions. Thiese 3 lines return the classification of the ensemnble j
            predK = predK[:l] #trim down for SMOTE
            for x in range(len(predK)):
                predK[x] = convert(predK[x])
            with np.errstate(divide = 'ignore', invalid = 'ignore'): #avoid early stopping
                #print("Length preds, and predK: ", len(predictions), len(predK))
                score = metrics.cohen_kappa_score(predictions, predK)
                #print("SCORE: ", score)
                if np.isnan(score): scores.append(0.0)    #avoid out of bounds behaviour
                else: scores.append(score)
    
    return min(scores) #the least agreeing classifier

def reference_vector(i, ensemble_probs, truth): #compares how close individual predictions are to entire ensemble prediction by the use of cosine similarity
    l = len(truth)
    x = ensemble_probs.mean(axis=0).argmax(axis=1)
    x = x[:l] #prun for SMOTE
    z = ensemble_probs[i,:].argmax(axis=1)
    z = z[:l] #prune for SMOTE
    ref = 2 * ( x== truth) - 1.0 
    ipred = 2 * (z == truth) - 1.0
    
    return 1.0 - scipy.spatial.distance.cosine(ref, ipred)
```

```python
def rank_pruning_measures(dictionary, data, label):
    #ensemble members dictionary
    MDM_list = []
    Weighted_MDM_list = []
    IC_list = []
    class_error_list = []
    neg_AUC_list = []
    error_ambiguity_list = []
    cohen_kappa_list = []
    reference_vector_list = []
    Votes = get_votes(data, dictionary)
    WVotes = get_votes_weighted(data, dictionary, label)
    ensemble_probs = get_ensemble_probs(dictionary)

    for i in range(len(dictionary)):      
        MDM_list.append(calc_MDM(Votes, dictionary[i]['classification'], dictionary[i]['true_class_label'], 0.2))
        Weighted_MDM_list.append(weighted_MDM(WVotes, dictionary[i]['classification'], dictionary[i]['true_class_label'], 0.2))
        IC_list.append(calc_IC(Votes, dictionary[i]['classification'], dictionary[i]['true_class_label']))
        class_error_list.append(cal_classifier_error(dictionary[i]['classification'], dictionary[i]['true_class_label']))
        neg_AUC_list.append(negative_AUC(dictionary[i]['classification'],  dictionary[i]['true_class_label']))
        error_ambiguity_list.append(error_ambiguity(i, ensemble_probs, dictionary[i]['true_class_label']))  #maybe wrap the ensemble probs in a function call
        cohen_kappa_list.append(kappa_statistic(i, ensemble_probs, dictionary[i]['classification'], dictionary[i]['true_class_label']))
        reference_vector_list.append(reference_vector(i, ensemble_probs, dictionary[i]['true_class_label']))
    
    measures_dictionary = {'MDM':MDM_list, 'Weighted MDM':Weighted_MDM_list, 'IC':IC_list, 'Class Error':class_error_list, 'Neg AUC':neg_AUC_list, 'Error Ambiguity':error_ambiguity_list, 'Cohen Kappa':cohen_kappa_list, 'Reference Vector':reference_vector_list}

    return measures_dictionary
    
def rank_pruning(ensemble, dictionary, sub_ensemble_size, key): #measure dictionary is input
    pruned_ens = ensemble.copy() #so we don't overwrite the input DF
    measures_df = pd.DataFrame(dictionary)
    pruned_ens = pd.concat([pruned_ens, measures_df], axis=1)  #for use in plotting the performance later
    pruned_ens[key] = dictionary[key]
    pruned_ens = pruned_ens.sort_values(by = key, ascending = False) 
    #pruned_ens = pruned_ens.sort_values(by = 'cohen_kappa', ascending = False) #sort ascending for cohen's kappa keeps most disagreeng 
    pruned_ens = pruned_ens.iloc[:sub_ensemble_size]
    
    #now getting the pruned ensemble for trianing input
    pruned_trainer = pruned_ens.loc[:, ['threshold', 'feature', 'direction', 'alpha', 'error']]
    
    return pruned_ens, pruned_trainer

def rank_pruning_process(dictionary, data, label, ensemble, sub_ensemble_size, key): #ensemble members dict
    #enahnce to optimise for best ensemble give measure passed in as key. key var is novel metric for ranking
    measure_dict = rank_pruning_measures(dictionary, data, label)
    pruned_ens, pruned_trainer = rank_pruning(ensemble, measure_dict, sub_ensemble_size, key)
    train_rank_pruned_df = predict_ensemble(pruned_trainer, data, label)
    
    #rp_cdf has all the measures attached, rp_tdf has the pruned df, train_rank_pruned_df is the final model after pruning
    return train_rank_pruned_df, pruned_ens, pruned_trainer
```

```python
#I want to find the best score given all possible combinations of features, so we use this function
def recursive_feat_select(features, X_train, y_train, X_test, y_test): #give X_train & X_test as DF
    lr = LogisticRegression(class_weight='balanced', solver='lbfgs', random_state=42, n_jobs=-1, max_iter=500)

    # Create RFE object and fit it to the data
    rfe = RFE(lr, n_features_to_select= features )
    rfe.fit(X_train, y_train)

    rfe_pred = rfe.predict(X_test)
    acc = metrics.accuracy_score(y_test, rfe_pred)
    #print(f"Test Accuracy with {features} features: ", acc)
    
    return acc, rfe.support_

def model_rfe(X_train, y_train, X_test, y_test):
    scores = []
    best_score = 0
    best_features = np.zeros(len(X_train.columns))
    
    for i in range(len(X_train.columns)):
        score, selected_features = recursive_feat_select(i+1, X_train, y_train, X_test, y_test) #i+1 to avoid 0 features
        scores.append((i+1, score))
        if score > best_score: 
            best_score = score
            best_features = selected_features

    return scores, best_score, best_features
```

```python
def calibration_disp(fit, pred, name):
    prob_true, prob_pred = calibration_curve(fit['true_class_label'], fit['pred_prob_pos'], n_bins=30)

    #calibrating training data
    prob_true_cal, prob_pred_cal = calibration_curve(fit['true_class_label'], fit['cal_prob'], n_bins=30, normalize=True)

    #TEST DATA
    #uncalibrated test data
    prob_true_test, prob_pred_test = calibration_curve(pred['true_class_label'], pred['pred_prob_pos'], n_bins=30)

    #calibrate test data
    prob_true_cal_test, prob_pred_cal_test = calibration_curve(pred['true_class_label'], pred['cal_prob'], n_bins=30, normalize=True)
    
    bscore1_uncal = metrics.brier_score_loss(fit['true_class_label'], fit['pred_prob_pos'])
    bscore1_cal = metrics.brier_score_loss(fit['true_class_label'], np.where(fit['cal_prob'] < 0, 0, fit['cal_prob']))

    bscore1_uncal_test = metrics.brier_score_loss(pred['true_class_label'], np.where(pred['pred_prob_pos'] < 0, 0, pred['pred_prob_pos']))
    bscore1_cal_test = metrics.brier_score_loss(pred['true_class_label'], np.where(pred['cal_prob'] < 0, 0, pred['cal_prob']))
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    ax1 = plot_calibration_curve(prob_true, prob_pred, 30, "Uncalibrated", bscore1_uncal, ax = ax1)
    ax2 = plot_calibration_curve(prob_true_cal, prob_pred_cal, 30, "Calibrated", bscore1_cal, ax = ax2)
    plt.suptitle(f"Training Data {name}", size =20)
    plt.tight_layout()
    plt.show()
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    ax1 = plot_calibration_curve(prob_true_test, prob_pred_test, 30, "Uncalibrated", bscore1_uncal_test, ax = ax1)
    ax2 = plot_calibration_curve(prob_true_cal_test, prob_pred_cal_test, 30, "Calibrated", bscore1_cal_test, ax = ax2)
    plt.suptitle(f"Test Data {name}", size =20)
    plt.tight_layout()
    plt.show()

    return
```

## 1.2 Explore margins
[Back to top](#Index)  | [To section head](#Task-1:-ADABOOST-IMPLEMENTATION) |  [To template section](#2.-Explore-Margins-and-Analyse-Results:-(5-marks))

First, we view the margin Rorschach plot below. Showing that the model is very confident in the corner reggions and the direct centre of the sample space. Which makes sense given the linear decision boundaries of AdaBoost. After 30 terations the model has divided the space into a grid, classifying data by class for given grid spaces. 

Below this we investigate the margins for models with 5, 10, 30 and 100 iterations. Here we see that as iterations increase confidence in model predictions increases, with diminishing returns. This is shown by the area under the margin curve being primarily above 0.5, the decision threshold. From 30 members to 100 members the improvement is negligible. At least for the simple, synthetic dataset. 

When different proportions of the sample are misclassified by 5%, 10%, 15%, 30% and 50% the margin curves are less confident shown by the increasing area between 0 to 0.5. 30% adn 50% are primarily unconfident in it's classificaiton. while 5% has no impact and 10% and 15% have increasing impact. 
Which is expected. The whole premise of AdaBoost is that it is rewarded for positive classifications. When the data is purposely misclassified the model gets confused. It has learned a pattern for one class, but up to 50% of the sample doens't follow the pattern. And so can't learn. 
Put simply, it is getting punished for no reason. It may have learnt the correct label, but the datset has the wrong one. AdaBoost's weight update goes awry resulting in lower alpha scores for each ensemble member. Visually represented in more of the margin curve area being below 0.5.

Of course, correctly labelled data is best. However, the implications of this in the real world, where data quality is often poor, is that AdaBoost will suffer where data quality is bad, but there is an allowable amount of misclassification. Here, 5% or less.

For our first experiment I replaced the decision stump with a logistic classifier. This classifier is no good, as can be seen from the plots below. The logistic learner clips the data to within randomly selected thresholds, then does a slogistic regression on the data within the reduced set. I took this approach because the logistic learner is stronger than a decision stump, and if the whole data set is used it will find a local minima at the edge of the space. This tendency is due to the graident descent of logistic regression. And, even with the randomised inputs, the model has this same tendency. As a result of the random subsets of the data, the error term in the output is misleading, as it is the lowest error of the subsets classified. 
Note, that the model can get confused and worsen the error, as with the ensemble with 50 iterations. This shows that the model is trying to learn, it just can't learn the data well, and so gets confused. 

The result is that the model learns that most of the data are of class -1, and so minimises the error by smple classfying most points as -1. With an error around 0.4 to 0.5. Slightly better than 50/50, but only because the classes are imbalanced. 

Why this occurs is because logistic regression classifies each point to -1 or 1, and so when the points change so does the classification. It is susceptible to variation, and more so when the classes are not linearly seperable, given that logistic regression is, fundamentally, a linear classifier.

### 1.3.2 Decision Tree classifier
[Back to top](#Index)          | [To section head](#Task-1:-ADABOOST-IMPLEMENTATION)  | [To tree classifier code](#Decision-tree-classifier)   |  [To template section](#3.-Adjust-Weak-Classifier-Type:-(5-marks))

Next I implemented a decision tree of varying depths as the weak learner. In this experiment we observe that the decision tree is also less good than a simple decision stump in general. our test AdaBoost with 30 iterations achieved an error of 0.00625 wiht one stump. Our AdaTree requried 100 iterations and a depth of 15 to get the same result. 

For the purpose of this implementation I didn't use the direction and threshold classification of AdaBoost, as the decision tree has many of each, and after testing a few AdaTrees I saw the performance was poor, and so didn't see the need to spend more time getting the implementation perfect. 

To explain what we see with a decision tree, this poor performance is due to the decision tree having inherently high variance and low bias as it become complex (deeper). The core idea of AdaBoost is to combine weak learners and iteratively get better. When we try 30 iterations with depth 2, the error is 0.22. At depth 5 the error is 0.17. One would expect better performance. 

Here, the tree with its high variance and low bias is fitting to the noise and becmoing confused. Doing the exact opposite of AdaBoost, which relies on stable weak learners that each have low variance, even if poor. SMOTE resampling works best when there are unevenly distributed classes, which is the case with the synthetic data. There are 103 cases of -1 and 57 for 1. 

The implementation of SMOTE here iteratively adds n new samples from k neighbours with w weight at the end of each tuning round. Up until the classes are even. If the minorty class becomes the majority, the samples generated will swap to the new minority class. To demonstrate this process working, the below graph circles the SMOTE re-sampled data points in black. 

```python
new_data, new_labels = SMOTE(X_train, y_train,  20, 3)

#add new labels to test_label 
SMOTE_data = np.append(X_train, new_data, axis=0)
SMOTE_label = np.append(y_train, new_labels)

x = new_data[:, 0]
y = new_data[:, 1]

pos1 = np.nonzero(SMOTE_label == 1)
pos2 = np.where(SMOTE_label == -1)

plt.figure()
plt.plot(SMOTE_data[pos1, 0], SMOTE_data[pos1, 1], 'o',c='gold', markersize=5)
plt.plot(SMOTE_data[pos2, 0], SMOTE_data[pos2, 1], 'o',c = 'teal', markersize=5)
plt.plot(x, y, 'o', markersize = 10, markeredgecolor='black', markerfacecolor='none')

plt.axis([-4.1, 4.1, -4.1, 4.1])
plt.title("SMOTE samples in training data", size=20)

plt.show()
```

First, when we use SMOTEBoost (AdaBoost with SMOTE resampling) to generate 2 samples from 10 neighbours at weight 0.005, we see the model performs worse than a regular model. This we can see from the classification of the evaluation data, where an entire central band is calssified as 1. Why this happens is because the parameters are not tuned. SMOTEBoost is creating a bunch of samples at the edges. More samples created nearby will add confusion to the model. It will add more data points which have less variability amongst them. So we are adding unbeneficial variability to the data and buying bias, which skews the bias varaince tradeoff. We can tuned this for better perfromance. 

```python
SMOTE_params = {'SMOTE' : True, 'new samples' : 2, 'neighbours' : 10, 'new weights' : 0.005}
SB_ens, SB_fit = Adaboost(10, X_train, y_train, SMOTE_params, None, False) #more nieghbours seems to be better, but SMOTE is random...
```

When we increase the number of neighbours and reduce the weight of the new points, then the model performs better than regular AdaBoost. Error of 0.158 vs regular error of 0.166.
This works better because the newly generated points are less like each other, while still being of the same class. We are now adding beneficial variance. We are also not weighting them too highly, rather we are sneaking them in. Now, the weight will need to be higher for more complex data sets if we want the next learning round to focus more on the new points, but here with a weight of 0.0001, the new points sneak on in and gently influence the model. 

This exercise shows that we can tune the model to perform better or worse. 

In general,  SMOTE adds noise to the model because the resampling occurs after the weight normalisation. In fact it doens't matter whether we invoke SMOTE before or after the weights, the result is the same. Because they get normalised in the next boosting round regardless. 

The introduction of new information to the model after it has trained a learner on the data is what confuses the votes at the end. There is more error introduced into the fitting. Hence the need for tuning. As we see, less weight to new points (introducing less noise) with more neighbours to identify likeness produces a model with only slightly better performance than regular Adaboost. This is because less weighted new points are 'quietly' slipping into the algorithm and balancing the minority class. Increasing the count of data points in this class. With our small dataset here, adding too many new points will add too much bias and the model will be confused too fast and too soon. After 10 iterations 20 new samples are in the minoirty class. These 20 samples are gradually learned, 'nudging' the model to improve it's predicitons. 

```python
SMOTE_params = {'SMOTE' : True, 'new samples' : 2, 'neighbours' : 20, 'new weights' : 0.0001}
SB_ens, SB_fit = Adaboost(10, X_train, y_train, SMOTE_params, None, False) #more nieghbours seems to be better, but SMOTE is random...SMOTE_predictions = predict_Adaboost(X_test, SB_ens, SB_fit)

SMOTE_predictions = predict_Adaboost(X_test, SB_ens, SB_fit)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
ax1 = plot_predict(X_test, SMOTE_predictions, y_test, 'Classified Data',ax= ax1)
ax2 = plot_predict(eval_data, predict_Adaboost(eval_data, SB_ens, SB_fit), eval_label, 'Evaluation Data',ax= ax2)
plt.suptitle("SMOTE Boost 10 iterations, 2 samples, 20 neighbours & weight 0.0001", size = 25)
plt.subplots_adjust(top=0.8)
plt.show()
```

### 1.4.2 Difficult examples: Cost sensitive boosting
[Back to top](#Index)          | [To section head](#Task-1:-ADABOOST-IMPLEMENTATION) | [Cost adjustment functions](#SMOTE-and-cost-functions)  |  [To template section](#4.-Focus-on-Difficult-Examples-/-Adjust-Classification-Threshold:-(5-marks))

With AdaCost the story is different. AdaBoost learns faster because it gets reinforcement about it's errors from the first boosting round. Making the model converge very fast. We see this with the low error per ensemble member. 

AdaCost has a simple premise. Increase the weighing process for misclassified data points to force the model's attention to shift faster. In other models this may be called learning rate. In essence, we are increasing the strength of the reinforcement on AdaBoost. 
The cost adjustment penalty formula used is from Fan et al (1999):
   
   $$D_{t+1}(i) = \frac{D_{t}(i) e^{-\alpha_{t}y_{i}h_{t}\frac{x_{i}}{\beta(i)}}}{Z_t}$$


Depending on the dataset complexity, more or less penalty will be needed. In general terms, a higher penalty means faster learning, and is required for very complex data. While a lower one results in lsower learning, which suits less complex data. Mainly because AdaBoost works on weak learners. AdaCost mkaes the learners less weak, and the impact of expert learners is stronger on less complex data. 

On this data the error reduces to 0.128 with  penalty of 0.5 or 0.9 on the evaluation data. What is of interest is the performance on the test data. a penalty of 0.5 results in some poor classification on the test set, seen by the increased numbner of misclassified point in the top middle of the space. While with a penalty of 0.9 the test data is much more central. On the evaluation data the final result is the same. Quite simply because with ten iterations there is a limit to how much can be learnt and even with a high penalty, the model hasn't learnt all it can by this point. 

```python
COST_params = {'COST' : True, 'cost_penalty' : 0.5}

cost_ens, cost_fit = Adaboost(10, X_train, y_train, None, COST_params, False)
cost_ens

#holy crap, way improved!
```

### 1.4.5 Classification Thresholds: Calibration
[Back to top](#Index)   | [To section head](#Task-1:-ADABOOST-IMPLEMENTATION) | [Calibration code](#Calibration-function)  |  [Calibration plot code](#Calibration-plots)  |    [To template section](#4.-Focus-on-Difficult-Examples-/-Adjust-Classification-Threshold:-(5-marks))

Calibration is another way to alter the decision making behaviour of AdaBoost. According to Haung et al (2020) for AdaBoost, isotonic regression is the way to go as AdaBoost tends to produce extreme probability estimates close to 0 and 1 and isotonic regression handles extreme better than Platt scaling. 

For ease, this calibration has been built into the AdaBoost implementation using the following formulas to get probabilities:

   $$ P(y=1|x) = \frac{1}{1 + e^{-2 * F(x)}}$$
   
   $$ F(x) = \sum^T_{t=1} \alpha_t H_t (x) $$
   
The effect of calibration on the Adaboost model is to  decrease the Brier score from 0.015 to 0.0063 - an almost perfect calibration on training data. And from 0.0668 to 0.075 on test data (slightly worse). The decision bins are pretty much all at either 0 or 1 (-1 and 1). 

We do see that calibrating this data worsens performance on the test data just slightly, which is expected when calibrating a model. Here the decrease is not much, but still not an improvement.

### 1.5.2 Rank pruning using novel ranking measures
[Back to top](#Index)          | [To section head](#Task-1:-ADABOOST-IMPLEMENTATION)  |  [To Rank refinement code](#Rank-refinement-pruning)  |    [To template section](#5.-Implement-an-ensemble-pruning/optimisation-procedure-that-is-executed-after-the-model-training-:-(5-marks))

To rank prune requires a measure for ranking. Initially I wanted to use the Cohen kappa measure to treat each ensemble member like a doctor independently performing the task and to measure the level of agreement between them. In researching this I found many other measures to try out. These are:

* Margin Diversity (Guo et al, 2018) where each member is ranked by its contribution to the margin. Designed for algorithms like AdaBoost. 
* Individual contribution (Lu et al, 2010) where each ensemble is raned by its contribution to the prediction with respect to the entire ensemble. 
* Individual error (Jiang et al, 2017) which is the average error of the ensemble member.
* Error ambiguity (Jiang et al, 2017) where each classifier is ranked according to the ambiguity decomposition - very similar to mean squared error calculated from the prediction probability.
* negative AUC score
* individual kapa statistic (Margineantu et al, 1997) which is the cohen kappa ranking for each ensemble. 
* reference vector (Hernandez et al, 2006) which is ranking by cosine simularity of a member to the entire ensemble's prediction. 

In this task we prune the 30 iteration AdaBoost model. We see that for a given metric measures perform differently. This is a game of picking the metric to optimise and taking the number of members in the ensemble where the metric peaks. For example, if we want to optimise the accuracy metrci then the negative AUC score does this with 27 members. And for weighted margin diversity (weighted MDM) this occurs at 26 ensemble members.

First of all, Adaboost needs numeric values. There are ordinal loan grades ranging from A to G. A is better, so we will codes these from 6 to 0. Since G is pretty much never ever getting a loan. 
Then there are categorical loan intents and home ownership. These have no implied order and so will be dummy coded as 1 or 0. 
Finally an individua's age can't be less than their employment length, which I noticed when glaning throiugh the data, so we will ensure that these values are removed. 
Aside from this, no other feature engineering is going to be done. 

```python
def kaggle_feat_enginer(df):
    #train = False
    
    #if 'loan_status' in df.columns: train = True
        
    df.drop((df[df['person_age'] <= df["person_emp_length"]].index))

    #change {Y, N} values to {1, 0}
    df['cb_person_default_on_file'].replace({'Y': 1, 'N': 0}, inplace = True)
    df['loan_grade'].replace({'A': 6, 'B': 5, 'C': 4, 'D': 3,'E': 2, 'F': 1, 'G': 0}, inplace = True)

    #encode categorical variables
    df = df.join(pd.get_dummies(df['person_home_ownership']))
    df = df.join(pd.get_dummies(df['loan_intent']))
    
    #change label for Adaboost
    if 'loan_status' in df.columns:
        df['loan_status'].replace(0, -1, inplace = True)
    #make label dataset
        df_label = np.array(df["loan_status"])
        #remove unnecessary data
        df.drop(['id', 'loan_status', 'loan_intent', 'person_home_ownership'], axis=1, inplace=True)

        return df, df_label
    else: 
        df.drop(['id', 'loan_intent', 'person_home_ownership'], axis=1, inplace=True)
        return df
```

MCC score is chosen because it is sensitive to the threshold choice and shows a balance measure of binary classification quality. We will use the bar chart ot get an idea of what values to use.
Which for the SMOTEBoost model is around 0.59 and 0.6 (on the left). For the SMOTECost model (on the right) is between 0.567 to 0.568.

When we try 100 threshold between 0.59 and 0.6, for SMOTEBoost, we find that the best thrshold is still 0.5. And the same of SMOTECost. Why? well, this is likely because the optimal threshold for this model really is 0.5. 

However, when we look at MCC we get threshold of 0.6and 0.56719 respectively. So we shall try these and see if they imporve the submissiom score.

```python
#Kada_01_ALL_fit
#Kada_11_ALL_fit
#K_fit
#tweak_plot(Kada_11_ALL_fit, y_train [0.4,0.5])
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 6))

ax1 = tweak_plot(Kada_11_ALL_fit, Ky_train, np.arange(0.59, 0.65, 0.01), ax1)
ax2 = tweak_plot(GS_test30_pred, test_label, np.arange(0.56, 0.57, 0.001), ax2)
#ax3 = tweak_plot(GS_test30_pred, test_label, [0.55, 0.56, 0.57, 0.58, 0.59], ax3)
fig.suptitle("SMOTEBoost on the left, SMOTECost on the right", size = 20)
fig.show()
```

```python
def tune_threshold(fit, thresholds, score):
    p = model_performance(fit)
    best_score = p[score]
    probs = fit["probabilities"]
    best_thresh = 0.5
    best_preds = fit['classification']

    for i in range(len(thresholds)):
        new_df = fit[['classification', 'true_class_label', 'probabilities', 'pred_prob_pos', 'cal_prob']]
        new_df['classification'] = threshold_predict(probs, thresholds[i])
        #print("PREDDS" ,new_df, "\n")
        #get the new score
        
        new_p = model_performance(new_df)
        new_score = new_p[score]
        if new_score > best_score: 
            best_score = new_score
            best_thresh = thresholds[i]
            best_preds = new_df['classification']       
    
    return best_thresh, best_score, best_preds
```

```python
bt_11, bs, bp = tune_threshold(Kada_11_ALL_fit, np.arange(0.59, 0.6, 0.001), "MCC")
print("BEST THRESHOLD: ", bt, "\nBEST MCC SCORE: ", bs)
```

```python
bt_01, bs, bp = tune_threshold(Kada_01_ALL_fit, np.arange(0.567, 0.5681, 0.0001), "MCC")
print("BEST THRESHOLD: ", bt, "\nBEST MCC SCORE: ", bs)
```

```python
def get_model_metrics(y_test, preds, probs):
    results = pd.DataFrame([np.array(y_test), preds, probs[:,1], probs[:,0]])
    results = results.T
    results.columns =  ['true_class_label', 'classification', 'pred_prob_pos', 'pred_prob_neg' ]
    results['probabilities'] = results['pred_prob_pos']
    results['true_class_label'] =  results['true_class_label'].replace(0, -1)
    results['classification'] =  results['classification'].replace(0, -1)
    results['cal_prob'] = isotonic_calibration(results['true_class_label'], results['pred_prob_pos'])
    results['cal_preds'] = np.sign(results['cal_prob'])

    Scores = model_performance(results)
    return Scores
```
