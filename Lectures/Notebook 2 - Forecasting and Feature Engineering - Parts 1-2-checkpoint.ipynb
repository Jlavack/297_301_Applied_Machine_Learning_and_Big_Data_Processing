# Univariate vs. Multivariate Forecasting

## Univariate Forecasting

### Definition:
- **Univariate Forecasting** involves predicting future values based on a single time series. The model uses historical data of only one variable to make forecasts.

### Characteristics:
- **Simplicity**: Easier to implement and interpret.
- **Limited Contextual Information**: Lacks data from other sources, which can restrict predictive performance, especially in dynamic environments.
- **Common Techniques**: ARIMA, Exponential Smoothing, Simple Moving Averages.

### Example:
- Forecasting future sales of a product using only past sales data.

## Multivariate Forecasting

**Definition**  
Multivariate forecasting uses multiple input time series to predict one or more target variables. This includes both:

- **Multiple Inputs → Single Output (MISO)**: Use several predictors to forecast one target variable.
- **Multiple Inputs → Multiple Outputs (MIMO)**: Forecast multiple target variables simultaneously.

> Note: "Multivariate" refers to the dimensionality of the input feature space, not necessarily to multiple outputs.

**Key Properties**
- **Richer Feature Set**: Uses lagged values of both the target and additional variables.
- **Increased Model Complexity**: More demanding in terms of data preparation and model design.
- **Common Methods**: VAR, multivariate regression, random forests, gradient boosting, kernel methods, LSTMs, TCNs, transformers.

**Example**  
Forecasting product sales using historical sales, marketing expenditure, and macroeconomic indicators.


## Covariates and Exogenous Variables

**Definition**  
Covariates refer to auxiliary input variables used to improve forecast accuracy. These can include:

- **Endogenous Variables**: Influenced by or jointly determined with the target variable.
- **Exogenous Variables (X or XREG)**: External variables not affected by the target variable, assumed to be independent or fixed during forecasting.

**Distinction from General Multivariate Forecasting**
- Multivariate forecasting includes any model with more than one input time series.
- The term "with exogenous variables" usually refers to multivariate forecasting where the additional inputs are external and not predicted jointly.

**Example**
- **Endogenous multivariate setup**: Forecasting GDP, inflation, and interest rate jointly using a VAR model.
- **Multivariate with exogenous variables**: Forecasting GDP using oil prices and temperature, treating these as fixed inputs.



This transforms the problem into a supervised learning task with a multivariate input space.

# Concept of Forecast Horizon

### Definition:
- The **Forecast Horizon** refers to the length of the future period over which predictions are made. It specifies how far into the future the forecast extends.

### Short-Term vs. Long-Term:
- **Short-Term Forecasting**: Typically involves a forecast horizon of days to weeks. It is often used for operational and tactical decisions.
- **Long-Term Forecasting**: Involves a forecast horizon of months to years. It is used for strategic planning and long-term decision-making.

### Impact on Model Choice:
- The forecast horizon can significantly influence the choice of model and its performance. Short-term models may focus on capturing immediate trends and patterns, while long-term models need to account for more substantial variability and uncertainty. Long-term forecasts need to rely more on seasonal and cyclical patterns as well as trends if they exist in the data in order to have utility.

# Converting Univariate to Multivariate Forecasting Using Lagged Values and Exogenous Variables

### Lagged Values:
- **Lagged Values** refer to the past observations of a variable that are used as inputs to forecast future values. By including lagged values as additional variables, a univariate time series can be transformed into a multivariate time series.

### Exogenous Variables:
- **Exogenous Variables** (external factors) are additional variables that are believed to influence the variable of interest but are not influenced by it. These variables can provide valuable information that improves the accuracy of forecasts.

### Process:
1. **Identify Lags**: Determine the number of lagged values to include based on the autocorrelation structure of the time series.
2. **Identify Exogenous Variables**: Select external factors that are relevant to the forecasting task.
3. **Create Lagged Variables**: Generate new variables from the original time series, where each new variable represents a lagged version of the original series.
4. **Model Building**: Use these lagged variables and exogenous variables as inputs in a multivariate forecasting model.

### Example:
- Original Time Series: $( y_t $)
- Lagged Values: $( y_{t-1}, y_{t-2}, \ldots, y_{t-p} $)
- Exogenous Variables: $( x_{1,t}, x_{2,t}, \ldots, x_{k,t} $)
- Multivariate Model: $( y_t = f(y_{t-1}, y_{t-2}, \ldots, y_{t-p}, x_{1,t}, x_{2,t}, \ldots, x_{k,t}) $)

### Advantages:
- **Captures Dynamics**: Including lagged values helps the model capture temporal dynamics and dependencies in the data.
- **Incorporates External Information**: Exogenous variables can provide additional information that enhances the model's accuracy.
- **Improves Accuracy**: This approach can improve forecasting accuracy by providing more comprehensive information about the past behavior and influencing factors.

### Example in Practice:
- Suppose we have monthly sales data $( y_t $). We can create lagged variables $( y_{t-1} $) (sales last month) and $( y_{t-2} $) (sales two months ago), and include exogenous variables like marketing expenditure $( x_{1,t} $) and economic indicators $( x_{2,t} $) to build a multivariate forecasting model.

# Formulating a Time Series Problem as a Machine Learning Forecasting Problem

Formulating a time series problem as a machine learning forecasting problem presents several challenges, such as handling temporal dependencies, seasonality, and trends. Here are the three main approaches, along with a key additional approach:

## 1. Forecasting Absolute Values

### Description:
- This approach focuses on predicting the future absolute values of the target variable using historical data.
- The model is trained to output the direct future values without transformation.

### Example:
- Predicting monthly sales figures for the next year based on past sales data.

### Challenges:
- **Temporal Dependencies**: Requires careful feature engineering to capture temporal dependencies, such as autocorrelation.
- **Non-Stationarity**: Must handle non-stationarity in the data, including trends and seasonality, which can complicate the modeling process.
- **Model Complexity**: Selecting an appropriate model complexity to avoid overfitting or underfitting.

## 2. Forecasting Changes in the Target Variable

### Description:
- This approach predicts the change or difference in the target variable from one time point to the next, rather than the absolute value.
- It is particularly useful for capturing short-term variations and reducing the impact of trends.

### Example:
- Predicting the daily change in stock prices rather than the absolute price.

### Challenges:
- **Differencing**: Requires transforming the target variable into differences, which can introduce additional noise.
- **Volatility**: Changes can be more volatile and harder to predict accurately, necessitating robust modeling techniques.
- **Integration**: Ensuring that the predicted changes can be effectively integrated back to form absolute forecasts.

## 3. Modeling Seasonality, Cycles, and Trend

### Description:
- This approach decomposes the time series into its constituent components: trend, seasonal cycles, and residuals (noise).
- Separate models are built to forecast each component, and the results are combined for the final forecast.

### Example:
- Decomposing monthly sales data into a long-term trend, yearly seasonality, and random noise, then forecasting each component separately.

### Challenges:
- **Accurate Decomposition**: Requires robust methods to accurately decompose the time series.
- **Component Forecasting**: Each component must be forecasted independently, which can complicate the modeling process.
- **Recombination**: Ensuring that the recombination of forecasts from different components maintains consistency and accuracy.

## Putting It All Together

### Integrated Approach:
- The most effective forecasting models often integrate elements from all three approaches.
- A hybrid model might predict changes in the target variable while also incorporating features that capture seasonality and trend.

### Example:
- Predicting monthly sales by forecasting the differenced series (change in sales) and adding back the seasonal and trend components.

# Testing Approaches in Forecasting


## 1. Walk-Forward Rolling Validation

### Explanation:
- **Walk-Forward Rolling Validation** involves training the model on a fixed-size window of past data and then testing it on the next immediate period.
- After each iteration, the window "rolls" forward by the size of the testing period.

### Steps:
1. Train the model on an initial period (e.g., the first 12 months).
2. Test the model on the subsequent period (e.g., the next 6 months).
3. Move the training and testing windows forward by 6 months.
4. Repeat the process.

### Example:
- **Iteration 1**: Train on Jan-2021 to Dec-2021, test on Jan-2022 to Jun-2022.
- **Iteration 2**: Train on Jul-2021 to Jun-2022, test on Jul-2022 to Dec-2022.

### Purpose:
- Simulates real-time forecasting by continuously updating the model with new data, providing an ongoing assessment of model performance.

## 2. Expanding Window Validation

### Explanation:
- **Expanding Window Validation** involves starting with an initial training period and expanding the training set with each iteration to include more recent data.
- The test period remains fixed.

### Steps:
1. Train the model on an initial period (e.g., the first 12 months).
2. Test the model on the subsequent period (e.g., the next 6 months).
3. Expand the training window to include the previous test period and more recent data.
4. Repeat the process.

### Example:
- **Iteration 1**: Train on Jan-2021 to Dec-2021, test on Jan-2022 to Jun-2022.
- **Iteration 2**: Train on Jan-2021 to Jun-2022, test on Jul-2022 to Dec-2022.

### Purpose:
- Evaluates how the model's performance improves as more data becomes available for training, reflecting the model's adaptability to increasing information.

## 3. Blocked Time Series Cross-Validation

### Explanation:
- **Blocked Time Series Cross-Validation** divides the time series into several contiguous, non-overlapping blocks.
- Each block is used as a test set once, with the preceding blocks used as the training set.

### Steps:
1. Divide the data into blocks (e.g., 6-month blocks).
2. Train the model on the data up to the start of the current test block.
3. Test on the current block.
4. Repeat for each block.

### Example:
- **Iteration 1**: Train on Jan-2021 to Jun-2021, test on Jul-2021 to Dec-2021.
- **Iteration 2**: Train on Jan-2021 to Dec-2021, test on Jan-2022 to Jun-2022.

### Purpose:
- Provides a robust assessment of model performance across distinct time periods, ensuring the model generalizes well over different segments of the data.

## 4. Backtesting

### Explanation:
- **Backtesting** simulates how well the model would have performed historically by using historical data to predict future outcomes.
- It involves iteratively training and testing the model on different periods of historical data.

### Steps:
1. Train the model on historical data up to a certain point (e.g., the first 12 months).
2. Predict the outcomes for the next period (e.g., the next 6 months).
3. Move the training period forward or expand it to include more recent data.
4. Repeat the process.

### Example:
- **Iteration 1**: Train on Jan-2021 to Dec-2021, test on Jan-2022 to Jun-2022.
- **Iteration 2**: Train on Jan-2021 to Jun-2022, test on Jul-2022 to Dec-2022.

### Purpose:
- Validates the model's predictive performance by simulating real-world conditions, ensuring it uses only past data to predict future outcomes, avoiding data leakage and providing realistic performance metrics.

## Key Differences

- **Walk-Forward Rolling Validation**: Uses a fixed-size training window that rolls forward in time, simulating continuous updates and providing an ongoing performance assessment.
- **Expanding Window Validation**: Expands the training set over time, assessing how additional data impacts model performance and reflecting the model's adaptability to increasing information.
- **Blocked Time Series Cross-Validation**: Divides data into non-overlapping blocks, providing a comprehensive evaluation across distinct time periods, ensuring the model generalizes well over different segments.
- **Backtesting**: Simulates historical performance by iteratively training and testing on different periods, ensuring realistic predictions by using only past data for forecasting.

# Common Evaluation Metrics in Forecasting

Evaluating the performance of forecasting models is crucial to ensure their accuracy and reliability. Here are four of the most common and effective evaluation metrics used in forecasting:

## 1. Mean Absolute Error (MAE)

**Definition**:
- MAE measures the average magnitude of the errors between predicted and actual values, without considering their direction. It is the average of the absolute differences between forecasted and actual values.

**Formula**:
$ \text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i| $

**Pros**:
- Simple to understand and interpret.
- Provides a straightforward measure of the average forecast error.

**Cons**:
- Does not account for the direction of the errors (overestimation or underestimation).

## 2. Mean Squared Error (MSE)

**Definition**:
- MSE measures the average of the squared differences between predicted and actual values. By squaring the errors, it penalizes larger errors more than smaller ones.

**Formula**:
$ \text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $

**Pros**:
- Highlights larger errors due to squaring, making it useful for identifying significant mistakes.
- Commonly used in regression analysis and forecasting.

**Cons**:
- Sensitive to outliers because errors are squared, which can disproportionately affect the metric.

## 3. Root Mean Squared Error (RMSE)

**Definition**:
- RMSE is the square root of the MSE, providing an error metric in the same units as the original data. It combines the benefits of MSE but is easier to interpret.

**Formula**:
$ \text{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2} $

**Pros**:
- Easier to interpret than MSE because it is in the same units as the data.
- Emphasizes larger errors, highlighting significant discrepancies.

**Cons**:
- Like MSE, it is sensitive to outliers.

## 4. Mean Absolute Percentage Error (MAPE)

**Definition**:
- MAPE measures the average magnitude of the errors as a percentage of the actual values. It provides a relative measure of forecast accuracy.

**Formula**:
$ \text{MAPE} = \frac{1}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right| \times 100 $

**Pros**:
- Provides a relative measure of error, making it easy to interpret and compare across different datasets.
- Expresses error as a percentage, which is useful for understanding the scale of errors relative to the actual values.

**Cons**:
- Can be problematic if actual values are very small or zero, as it can produce extremely large or undefined values.

# Benchmarking Forecasting Models Against Baselines and Persistence Models

Benchmarking is a critical step in the forecasting process. It helps determine if the forecasting model provides meaningful improvements over simpler, baseline models or just persistence models (taking the previous value as the next). 

## Importance of Benchmarking

Benchmarking involves comparing the performance of your forecasting model against simpler models or benchmarks. The main objectives are:
- **Validation**: Ensure that the complex forecasting model performs better than simple alternatives.
- **Evaluation**: Identify areas where the model needs improvement.
- **Confidence**: Provide stakeholders with confidence in the model's predictions.

## Common Baseline Models

### 1. Naive Forecast (Persistence Model)

**Definition**:
- The **Naive Forecast** assumes that the future value is equal to the last observed value. It is the simplest form of forecasting and serves as a benchmark to beat.

**Formula**:
$ \hat{y}_{t+1} = y_t $

**Example**:
- If the sales in June were 200 units, the naive forecast for July would also be 200 units.

### 2. Seasonal Naive Forecast

**Definition**:
- The **Seasonal Naive Forecast** extends the naive forecast by accounting for seasonality. It assumes that the future value is equal to the last observed value from the same period in the previous season.

**Formula**:
$ \hat{y}_{t+s} = y_t $
where $ s $ is the seasonal period.

**Example**:
- If the sales in June last year were 180 units, the seasonal naive forecast for June this year would be 180 units.

### 3. Simple Moving Average

**Definition**:
- The **Simple Moving Average** forecast uses the average of a fixed number of the most recent observations.

**Formula**:
$ \hat{y}_{t+1} = \frac{1}{n} \sum_{i=0}^{n-1} y_{t-i} $

**Example**:
- If the sales for the past three months were 190, 200, and 210 units, the forecast for the next month would be the average: $(190 + 200 + 210) / 3 = 200$ units.

### 4. Exponential Smoothing

**Definition**:
- **Exponential Smoothing** forecasts future values by applying exponentially decreasing weights to past observations.

**Formula**:
$ \hat{y}_{t+1} = \alpha y_t + (1 - \alpha) \hat{y}_t $
where $ \alpha $ is the smoothing constant.

**Example**:
- If the smoothing constant $ \alpha $ is 0.5 and the last observed value was 210 units, the forecast might be a weighted average of the last observation and the previous forecast.

###  Time Series Decomposition and Simple Forecasting

Time series decomposition is a classical approach in time series analysis that separates a time series into distinct components, typically trend, seasonality, and residual (noise). This allows us to isolate long-term changes in level (trend), recurring intra-period patterns (seasonality), and irregular fluctuations (residuals). When a time series exhibits both a clear trend and a strong seasonal pattern, decomposition can provide a transparent foundation for simple forecasting. By independently extrapolating the trend and repeating the seasonal structure, we can construct a basic forecast without fitting a formal statistical model. This approach assumes that the trend continues at its most recent level and that the seasonal pattern is stable and repeats annually.

Although simplistic, this decomposition-based forecast can serve as a strong baseline in settings where interpretability is important or where model fitting is infeasible. This will hopefully help in building intuition about how each component contributes to the overall time series structure and forecast.

### Simple Decomposition-Based Forecasting

We construct an interpretable forecast using classical additive decomposition, where the trend is held constant at its most recent estimate, and the seasonal component is repeated from the same calendar month in the prior year. Residuals are assumed to be white noise and excluded from the forecast.

```python
# Get the last observed time step
# --- STEP 1: Prepare the time series ---
#Make sure the index is datetime and frequency is monthly
series = df_sum['total'].copy()

# --- STEP 2: Decompose the series ---
decomposition = seasonal_decompose(series, model='additive', period=12)

# --- STEP 3: Plot the decomposition ---
decomposition.plot()
plt.suptitle('Time Series Decomposition', fontsize=16)
plt.tight_layout()
plt.show()

# --- STEP 4: Forecast the next step using trend + seasonality ---
#Get the last timestamp and compute the next
last_timestamp = series.index[-1]
next_timestamp = last_timestamp + pd.DateOffset(months=1)

#Forecast trend using last available trend value
trend_series = decomposition.trend.dropna()
trend_forecast = trend_series.iloc[-1]

#Compute average seasonal pattern by month
seasonal_series = decomposition.seasonal.dropna()
seasonal_by_month = seasonal_series.groupby(seasonal_series.index.month).mean()

#Ensure all 12 months are present
seasonal_by_month = seasonal_by_month.reindex(range(1, 13), fill_value=0.0)

#Forecast seasonal component based on the next month
next_month_index = next_timestamp.month
#seasonal_forecast = seasonal_by_month.loc[next_month_index]
seasonal_forecast = decomposition.seasonal.loc[last_timestamp - pd.DateOffset(months=12)]

#Forecast = trend + seasonal (ignoring residuals)
forecast = trend_forecast + seasonal_forecast

print(f"Forecast for {next_timestamp.date()}: {forecast:.2f}")

# --- STEP 5: Plot the original series and forecast point ---
plt.figure(figsize=(10, 5))
plt.plot(series.index, series.values, label='Observed', color='blue')
plt.scatter([next_timestamp], [forecast], color='red', label='Forecast', zorder=5)
plt.axvline(x=series.index[-1], color='gray', linestyle='--', alpha=0.6)
plt.title(f"Forecast for {next_timestamp.strftime('%Y-%m')}: {forecast:.2f}")
plt.xlabel("Date")
plt.ylabel("Value")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import scipy.interpolate as inter

def fit_spline(x_col, y_col, df, predict_col):
    df_res = pd.DataFrame(pd.date_range(df.quarter.iloc[0], df.quarter.iloc[len(df)-1], freq='M'), columns=['date'])
    df_res = df_res.reset_index().set_index('date')
    df_res.rename(columns={'index':x_col}, inplace=True)
    df_res['y'] = df.set_index('quarter')[y_col]
    sns.set_style("whitegrid")
    plt.figure(figsize=(6 * 10, 16))
    plt.xticks(fontsize=30)
    plt.yticks(fontsize=30)
    plt.ylabel('GDP Change',fontsize=50)
    plt.title('GDP Function Fit (Spline) -- ' + predict_col ,fontsize=50)
    df = df[[x_col, y_col]].dropna()
    spline_fit = inter.InterpolatedUnivariateSpline (df[x_col].iloc[1:].values, df[y_col].iloc[1:].values, k=5)
    df_res[predict_col] = spline_fit(df_res.x)
    plt.scatter(df_res['2001-03':].index, df_res['2001-03':].y, s=500)
    plt.plot(df_res['2001-03':][predict_col], 'r')

    return df_res[[predict_col]]
```

## Rolling calculations

Rolling functions—like moving averages—are often used in time series to smooth out short-term noise and help highlight underlying trends. This can be useful for machine learning because it lets the model focus on more stable patterns instead of reacting to every small fluctuation.

However, there’s an important drawback. As the rolling window size increases, the output becomes increasingly delayed. This means the smoothed value at time t actually reflects data from the past up to t, and the signal gets "pushed right" in time. In practice, this creates a lag, which can be problematic if you're trying to detect sudden changes or shocks in real time—the model won’t "see" the shift until later.

To address this, more sophisticated smoothing methods exist. One particularly useful one is the Savitzky–Golay filter—often abbreviated as savgol_filter in Python’s SciPy library. Unlike standard rolling averages, the Savitzky–Golay filter fits a polynomial to the window, which allows it to smooth the data without shifting the signal. This makes it more responsive to abrupt changes while still reducing noise.

- in order to be able to incorporate *N* number of previous predictions into the current prediction, we also need to feed these inputs into the training algorithm

- this means that when the algorithm is training to map *x --> y*, *N* number of features in the vector *x* will actually be *ys*. ie $y_{t-1}, y_{t-2}, ... y_{t-N}$  

- we therefore need to create additional input features which represent a lag of *y* values

- during the prediction phase, we need to consistantly update in feed back into the feature vector predictions from the current classifier

# Correlation analysis

Traditional correlation techniques like Pearson, Kendall, and Spearman are often not informative for time series data due to their inherent assumptions and limitations. Pearson correlation measures linear relationships but assumes that data points are independent, which is not the case in time series where autocorrelation is common. This autocorrelation can inflate or deflate correlation coefficients, leading to misleading results. Similarly, Kendall's Tau, which measures ordinal associations, and Spearman's Rank Correlation, which assesses monotonic relationships, also assume data point independence. Time series data's temporal dependencies and trends violate these assumptions, making these correlation measures unreliable. Additionally, these methods do not account for the temporal structure and dependencies inherent in time series data, which are crucial for understanding time-dependent relationships.

### Xi Correlation

The newly released correlation method called the ξ (xi) approach, developed by Professor Sourav Chatterjee, offers a more robust way to measure correlation, particularly suited for non-linear and complex relationships. Unlike traditional correlation measures, ξ does not assume linearity or independence between observations. It captures both linear and non-linear dependencies and is resilient to the effects of outliers and non-stationarity, making it especially useful for time series data where these issues are prevalent.

Key Features of the ξ Approach:

    Flexibility: Can capture both linear and non-linear relationships.
    Robustness: Less sensitive to outliers and does not require assumptions of independence.
    Suitability for Time Series: Handles non-stationary data effectively, making it ideal for complex time series analysis.
    
>> Chatterjee, S. (2021). A new coefficient of correlation. Journal of the American Statistical Association, 116(536), 2009-2022.
>> https://www.tandfonline.com/doi/abs/10.1080/01621459.2020.1758115

```python
from scipy.stats import rankdata, norm

def xicor2(x, y, ties="auto"):
    x = np.asarray(x).flatten()
    y = np.asarray(y).flatten()
    n = len(y)

    if len(x) != n:
        raise IndexError(f"x, y length mismatch: {len(x)}, {len(y)}")

    if ties == "auto":
        ties = len(np.unique(y)) < n
    elif not isinstance(ties, bool):
        raise ValueError(f"expected ties either \"auto\" or boolean, got {ties} ({type(ties)}) instead")
    
    y = y[np.argsort(x)]
    r = rankdata(y, method="ordinal")
    nominator = np.sum(np.abs(np.diff(r)))

    if ties:
        l = rankdata(y, method="max")
        denominator = 2 * np.sum(l * (n - l))
        nominator *= n
    else:
        denominator = np.power(n, 2) - 1
        nominator *= 3

    statistic = 1 - nominator / denominator
    p_value = norm.sf(statistic, scale=2 / 5 / np.sqrt(n))

    return statistic, p_value
```

The $\xi$ Xi correlation method above measures the relationship between two variables by ranking one variable based on the sorted order of another, then calculating the differences in these ranks. It adjusts for ties and scales the differences to provide a robust measure of correlation, capturing both linear and non-linear dependencies. This method is particularly useful for time series data, where traditional correlation techniques fail due to temporal dependencies and non-stationarity, offering a more reliable assessment of complex relationships.

```python
def xicor(x, y, ties="auto"):
    x = np.asarray(x).flatten()
    y = np.asarray(y).flatten()
    n = len(y)

    if len(x) != n:
        raise IndexError(f"x, y length mismatch: {len(x)}, {len(y)}")

    if ties == "auto":
        ties = len(np.unique(y)) < n
    elif not isinstance(ties, bool):
        raise ValueError(f"expected ties either \"auto\" or boolean, got {ties} ({type(ties)}) instead")
    
    y = y[np.argsort(x)]
    r = rankdata(y, method="ordinal")
    nominator = np.sum(np.abs(np.diff(r)))

    if ties:
        l = rankdata(y, method="max")
        denominator = 2 * np.sum(l * (n - l))
        nominator *= n
    else:
        denominator = np.power(n, 2) - 1
        nominator *= 3

    statistic = 1 - nominator / denominator

    return statistic
```

### Prepare $y_{t-1}, y_{t-2}, ... y_{t-N}$ features

```python
def create_prev_predictor_columns(df, col_name, suff, prev_periods):
    for c in df.columns:
        if c.find('minus') > -1:
            df.drop(labels=[c],axis=1,  inplace=True)

    for i in range(prev_periods):
         df[col_name + suff + str(i+1)] = df[col_name].shift(i+1)

    return df
```

```python
# prev_periods will define the N 
prev_periods = 2

# predict_col is the name of the Y column 
predict_col = 'estimated_quarterly_percent_change'

# create $y_{t-1}, y_{t-2}, ... y_{t-N}$ features
df_input = create_prev_predictor_columns(df_sum, predict_col, '_minus', prev_periods)
df_input.tail()

#this function creates lagged periods for the column input. Use it to get lagged time periods to help with the prediction
```

# Forecasting 

We saw how we can at time $t$ predict $y_t$, given $x_t$. The process involves training $x_t$ against $y_t$, while incorporating into the $x$ input vector a fixe $N$ number of $y_{t-1}, y_{t-2}, ..., y_{t-N}$ values.   

Forecasting refers to predicting $m$ number of timesteps ahead of $y_t$, namely, $y_{t+m}$.

In order to do this, we have to reformulate the training in such a way that we train each $x_t$ against $y_{t+m}$, while incorporating into the input vector $y_{t+m-1}, y_{t+m-2}, ..., y_{t+m-N}$ values.
